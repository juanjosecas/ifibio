```python
import ast
import pandas as pd
import unicodedata

import ast
import unicodedata


def _quitar_acentos(texto):
    texto = str(texto)
    texto = unicodedata.normalize("NFKD", texto)
    return "".join(c for c in texto if not unicodedata.combining(c))


def estandarizar_authors(authors_str):
    try:
        authors = (
            ast.literal_eval(authors_str)
            if isinstance(authors_str, str)
            else authors_str
        )
    except Exception:
        return authors_str

    if not isinstance(authors, list):
        return authors_str

    for author in authors:
        apellido = author.get("LastName")
        iniciales = author.get("Initials")

        if apellido and iniciales:
            # Nos quedamos solamente con la primera inicial
            inicial = str(iniciales).strip()[0]

            # Quitamos acentos para evitar duplicados como Martínez N / Martinez N
            apellido_limpio = _quitar_acentos(apellido).strip()

            author["AuthorClean"] = f"{apellido_limpio} {inicial.upper()}"

    return str(authors)

df["Authors"] = df["Authors"].apply(estandarizar_authors)

display(df["Authors"])

# print una celda de ejemplo para verificar que se haya aplicado correctamente la función
print(df["Authors"].iloc[0])  # Imprime la primera celda de la columna "Authors" para verificar
```

```python
import ast
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# Preparar año y número de autores
# ============================================================

# PublicationDate está guardado como diccionario en forma de texto
pub_date = df["PublicationDate"].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) else {}
)

df["Year"] = pd.to_numeric(pub_date.apply(lambda x: x.get("Year")), errors="coerce")


# Authors está guardado como lista de diccionarios en forma de texto
authors = df["Authors"].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else [])

df["N_Authors"] = authors.apply(len)


# ============================================================
# 1. Evolución temporal del número total de publicaciones
# ============================================================

publications_year = df.dropna(subset=["Year"]).groupby("Year").size()

plt.figure(figsize=(10, 5))

plt.plot(publications_year.index, publications_year.values, marker="o")

plt.xlabel("Año")
plt.ylabel("Número de publicaciones")
plt.title("Evolución temporal de las publicaciones")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "evolucion_temporal_publicaciones.png"), dpi=300, bbox_inches="tight")
plt.show()


# ============================================================
# 2. Histograma + KDE del número de autores por paper
# ============================================================

from scipy.stats import gaussian_kde
import numpy as np

x = df.loc[df["N_Authors"] > 0, "N_Authors"].dropna()

plt.figure(figsize=(10, 5))

plt.hist(x, bins=range(int(x.min()), int(x.max()) + 2), density=True, alpha=0.6)

# KDE
kde = gaussian_kde(x)

x_grid = np.linspace(x.min(), x.max(), 500)

plt.plot(x_grid, kde(x_grid), linewidth=2)

plt.xlabel("Número de autores")
plt.ylabel("Densidad")
plt.title("Distribución del número de autores por publicación")

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "distribucion_numero_autores.png"), dpi=300, bbox_inches="tight")
plt.show()


# ============================================================
# 3. Número de autores por paper a lo largo del tiempo
# ============================================================

plt.figure(figsize=(10, 5))

plt.scatter(df["Year"], df["N_Authors"], alpha=0.4)

plt.xlabel("Año")
plt.ylabel("Número de autores")
plt.title("Número de autores por publicación a lo largo del tiempo")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "autores_por_publicacion_tiempo.png"), dpi=300, bbox_inches="tight")
plt.show()


# ============================================================
# 4. Promedio y mediana de autores por paper por año
# ============================================================

authors_year = (
    df.dropna(subset=["Year"]).groupby("Year")["N_Authors"].agg(["mean", "median"])
)

plt.figure(figsize=(10, 5))

plt.plot(authors_year.index, authors_year["mean"], marker="o", label="Promedio")

plt.plot(authors_year.index, authors_year["median"], marker="o", label="Mediana")

plt.xlabel("Año")
plt.ylabel("Autores por publicación")
plt.title("Tamaño de los equipos de autores a lo largo del tiempo")

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "promedio_mediana_autores.png"), dpi=300, bbox_inches="tight")
plt.show()


# ============================================================
# 5. Número acumulado de publicaciones
# ============================================================

publications_cumulative = publications_year.cumsum()

plt.figure(figsize=(10, 5))

plt.plot(publications_cumulative.index, publications_cumulative.values, marker="o")

plt.xlabel("Año")
plt.ylabel("Publicaciones acumuladas")
plt.title("Crecimiento acumulado de las publicaciones")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "publicaciones_acumuladas.png"), dpi=300, bbox_inches="tight")
plt.show()


# ============================================================
# 6. Número total de autores firmantes por año
# ============================================================

total_authors_year = df.dropna(subset=["Year"]).groupby("Year")["N_Authors"].sum()

plt.figure(figsize=(10, 5))

plt.plot(total_authors_year.index, total_authors_year.values, marker="o")

plt.xlabel("Año")
plt.ylabel("Número total de firmas")
plt.title("Cantidad total de autores firmantes por año")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "total_autores_por_año.png"), dpi=300, bbox_inches="tight")
plt.show()


# ============================================================
# 7. Máximo número de autores en un paper por año
# ============================================================

max_authors_year = df.dropna(subset=["Year"]).groupby("Year")["N_Authors"].max()

plt.figure(figsize=(10, 5))

plt.plot(max_authors_year.index, max_authors_year.values, marker="o")

plt.xlabel("Año")
plt.ylabel("Máximo número de autores")
plt.title("Tamaño máximo de los equipos de autores por año")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "max_autores_por_año.png"), dpi=300, bbox_inches="tight")
plt.show()


# ============================================================
# 8. Productividad anual vs tamaño promedio de los equipos
# ============================================================

temporal = pd.DataFrame(
    {"Publications": publications_year, "MeanAuthors": authors_year["mean"]}
).dropna()

plt.figure(figsize=(8, 6))

plt.scatter(temporal["Publications"], temporal["MeanAuthors"])

for year, row in temporal.iterrows():

    plt.annotate(
        int(year), (row["Publications"], row["MeanAuthors"]), fontsize=8, alpha=0.7
    )

plt.xlabel("Número de publicaciones por año")
plt.ylabel("Promedio de autores por publicación")
plt.title("Productividad vs tamaño de los equipos")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "productividad_vs_tamano_equipos.png"), dpi=300, bbox_inches="tight")
plt.show()


# guardar las tablas con estas informaciones en archivos CSV
publications_year.to_csv(os.path.join(OUTPUT_DIR, "publications_per_year.csv"))
authors_year.to_csv(os.path.join(OUTPUT_DIR, "authors_per_year.csv"))
publications_cumulative.to_csv(os.path.join(OUTPUT_DIR, "publications_cumulative.csv"))
```

```python
# ============================================================
# Número de autores con afiliación IFIBIO por paper
# ============================================================

import ast

authors = df["Authors"].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else [])

df["N_Authors_IFIBIO"] = authors.apply(
    lambda author_list: sum(
        1
        for author in author_list
        if any(
            (
                "IFIBIO" in (aff.get("Affiliation") or "").upper()
                or "INSTITUTO DE FISIOLOGÍA Y BIOFÍSICA BERNARDO"
                in (aff.get("Affiliation") or "").upper()
                or "HOUSSAY" in (aff.get("Affiliation") or "").upper()
                or "IFIBIO-HOUSSAY" in (aff.get("Affiliation") or "").upper()
            )
            for aff in author.get("Affiliations", [])
        )
    )
)

# Mostrar resultado
display(df[["PMID", "Title", "N_Authors", "N_Authors_IFIBIO"]])
# guardar la version final del dataframe con la columna N_Authors_IFIBIO en un archivo CSV
df.to_csv(os.path.join(OUTPUT_DIR, "all_pubmed_metadata_ifibio_with_N_Authors_IFIBIO.csv"), index=False)

# Scatter: número total de autores vs autores con afiliación IFIBIO

import matplotlib.pyplot as plt

plt.figure(figsize=(7, 6))

plt.scatter(df["N_Authors"], df["N_Authors_IFIBIO"], alpha=0.6)

plt.xlabel("Número total de autores")
plt.ylabel("Número de autores IFIBIO")
plt.title("Autores totales vs autores IFIBIO")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Relación entre autores IFIBIO y autores totales

df["Fraction_IFIBIO"] = df["N_Authors_IFIBIO"] / df["N_Authors"]


# Promedio anual
fraction_year = (
    df.dropna(subset=["Year", "Fraction_IFIBIO"])
    .groupby("Year")["Fraction_IFIBIO"]
    .mean()
)


# Plot
plt.figure(figsize=(10, 5))

plt.scatter(df["Year"], df["Fraction_IFIBIO"], alpha=0.25, label="Papers")

plt.plot(
    fraction_year.index,
    fraction_year.values,
    marker="o",
    linewidth=2,
    label="Promedio anual",
)

plt.xlabel("Año")
plt.ylabel("N_Authors_IFIBIO / N_Authors")
plt.title("Proporción de autores IFIBIO por publicación a lo largo del tiempo")

plt.ylim(0, 1.05)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
# ============================================================
# Extraer todas las filiaciones de todos los autores
# ============================================================

import ast
import pandas as pd
import matplotlib.pyplot as plt

authors = df["Authors"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


# Una fila por autor × filiación × paper
affiliation_records = []

for pmid, author_list in zip(df["PMID"], authors):

    for author in author_list:

        author_name = author.get("AuthorClean")

        for aff in author.get("Affiliations", []):

            affiliation = aff.get("Affiliation")

            if affiliation:

                affiliation_records.append(
                    {"PMID": pmid, "Author": author_name, "Affiliation": affiliation}
                )


affiliations_df = pd.DataFrame(affiliation_records)

print("Registros autor × filiación:", len(affiliations_df))
print("Filiaciones diferentes:", affiliations_df["Affiliation"].nunique())

display(affiliations_df)
```

```python
# ============================================================
# Estadísticas de filiaciones
# ============================================================

affiliation_stats = (
    affiliations_df.groupby("Affiliation")
    .agg(
        N_records=("PMID", "size"),
        N_papers=("PMID", "nunique"),
        N_authors=("Author", "nunique"),
    )
    .sort_values("N_papers", ascending=False)
    .reset_index()
)

display(affiliation_stats)

# save to CSV
affiliation_stats.to_csv("affiliation_stats.csv", index=False)
```

```python
# ============================================================
# Top 25 filiaciones por número de papers
# ============================================================

top_affiliations = affiliation_stats.head(25)
top_affiliations["Affiliation"] = top_affiliations["Affiliation"].str.slice(
    0, 25
) + top_affiliations["Affiliation"].apply(lambda x: "…" if len(x) > 25 else "")
plt.figure(figsize=(10, 9))

plt.barh(top_affiliations["Affiliation"], top_affiliations["N_papers"])

plt.xlabel("Número de publicaciones")
plt.ylabel("Filiación")
plt.title("Principales filiaciones presentes en las publicaciones")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Evolución mensual del número de publicaciones
# ============================================================

import ast
import pandas as pd
import matplotlib.pyplot as plt

# Extraer año y mes desde PublicationDate
pub_date = df["PublicationDate"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, dict) else {})
    )
)

df["Pub_Year"] = pd.to_numeric(pub_date.apply(lambda x: x.get("Year")), errors="coerce")

df["Pub_Month"] = pub_date.apply(lambda x: x.get("Month"))


# Convertir nombres de meses a número
month_map = {
    "Jan": 1,
    "Feb": 2,
    "Mar": 3,
    "Apr": 4,
    "May": 5,
    "Jun": 6,
    "Jul": 7,
    "Aug": 8,
    "Sep": 9,
    "Oct": 10,
    "Nov": 11,
    "Dec": 12,
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12,
}

df["Pub_Month"] = df["Pub_Month"].replace(month_map)

df["Pub_Month"] = pd.to_numeric(df["Pub_Month"], errors="coerce")


# Crear fecha año-mes
df["YearMonth"] = pd.to_datetime(
    {"year": df["Pub_Year"], "month": df["Pub_Month"], "day": 1}, errors="coerce"
)


# Contar publicaciones por mes
publications_month = (
    df.dropna(subset=["YearMonth"]).groupby("YearMonth").size().rename("N_Publications")
)


# Completar meses sin publicaciones con cero
full_months = pd.date_range(
    publications_month.index.min(), publications_month.index.max(), freq="MS"
)

publications_month = publications_month.reindex(full_months, fill_value=0)


# Plot
plt.figure(figsize=(14, 5))

plt.plot(publications_month.index, publications_month.values, linewidth=1.5)

plt.xlabel("Fecha")
plt.ylabel("Número de publicaciones")
plt.title("Evolución mensual de publicaciones")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()

# plot de barras de meses en los que mas se publicaron articulos
# se calcula el promedio de publicaciones por mes y se ordena de mayor a menor
monthly_avg = (
    df.dropna(subset=["Pub_Month"])
    .groupby("Pub_Month")
    .size()
    .rename("N_Publications")
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))
plt.bar(monthly_avg.index, monthly_avg.values, color="skyblue")
plt.xlabel("Mes")
plt.ylabel("Número de publicaciones")
plt.title("Meses con mayor número de publicaciones")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Cantidad de publicaciones según número de autores IFIBIO
# ============================================================

ifibio_authors_distribution = (
    df["N_Authors_IFIBIO"]
    .value_counts()
    .sort_index()
    .rename_axis("N_Authors_IFIBIO")
    .reset_index(name="N_Publications")
)

# Porcentaje
ifibio_authors_distribution["Percent"] = (
    ifibio_authors_distribution["N_Publications"] / len(df) * 100
).round(1)

display(ifibio_authors_distribution)


# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(9, 5))

plt.bar(
    ifibio_authors_distribution["N_Authors_IFIBIO"],
    ifibio_authors_distribution["N_Publications"],
)

plt.xlabel("Número de autores IFIBIO en el paper")
plt.ylabel("Número de publicaciones")
plt.title("Publicaciones según número de autores IFIBIO")

plt.xticks(ifibio_authors_distribution["N_Authors_IFIBIO"])

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Autores más prolíficos
# ============================================================

import ast
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

authors = df["Authors"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


# Una fila por autor y publicación
author_records = []

for pmid, author_list in zip(df["PMID"], authors):
    for author in author_list:
        name = author.get("AuthorClean")

        if name:
            author_records.append({"PMID": pmid, "Author": name})


authors_df = pd.DataFrame(author_records)


# Número de publicaciones por autor
author_stats = (
    authors_df.groupby("Author")
    .agg(N_Publications=("PMID", "nunique"))
    .sort_values("N_Publications", ascending=False)
    .reset_index()
)


display(author_stats.head(30))


# ============================================================
# Plot: 30 autores más prolíficos
# ============================================================

top_n = 10

top_authors = author_stats.head(top_n).sort_values("N_Publications", ascending=True)


plt.figure(figsize=(10, 10))

sns.barplot(
    data=top_authors,
    x="N_Publications",
    y="Author",
    hue="N_Publications",
    palette="viridis",
    legend=False,
)

plt.xlabel("Número de publicaciones")
plt.ylabel("")
plt.title(f"{top_n} autores más prolíficos")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Dos autores más prolíficos de cada año
# ============================================================

# Una fila por autor y publicación, agregando año
authors_year_df = authors_df.merge(df[["PMID", "Year"]], on="PMID", how="left")


# Número de publicaciones de cada autor por año
author_year_stats = (
    authors_year_df.dropna(subset=["Year"])
    .groupby(["Year", "Author"])
    .agg(N_Publications=("PMID", "nunique"))
    .reset_index()
)


# Seleccionar los 2 autores con más publicaciones de cada año
top2_year = (
    author_year_stats.sort_values(["Year", "N_Publications"], ascending=[True, False])
    .groupby("Year")
    .head(2)
    .copy()
)

top2_year["Year"] = top2_year["Year"].astype(int)

display(top2_year)


# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(14, 7))

sns.barplot(data=top2_year, x="Year", y="N_Publications", hue="Author", palette="Set2")

plt.xlabel("Año")
plt.ylabel("Número de publicaciones")
plt.title("Dos autores más prolíficos por año")

plt.xticks(rotation=45)
plt.legend(title="Autor", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Dos autores IFIBIO más prolíficos de cada año
# ============================================================

# Identificar autores con filiación IFIBIO
ifibio_records = []

for pmid, year, author_list in zip(df["PMID"], df["Year"], authors):

    for author in author_list:

        name = author.get("AuthorClean")

        is_ifibio = any(
            "IFIBIO" in (aff.get("Affiliation") or "").upper()
            for aff in author.get("Affiliations", [])
        )

        if name and is_ifibio:
            ifibio_records.append({"PMID": pmid, "Year": year, "Author": name})


ifibio_authors_df = pd.DataFrame(ifibio_records)


# Publicaciones por autor IFIBIO y año
ifibio_year_stats = (
    ifibio_authors_df.dropna(subset=["Year"])
    .groupby(["Year", "Author"])
    .agg(N_Publications=("PMID", "nunique"))
    .reset_index()
)


# Dos más prolíficos por año
top2_ifibio_year = (
    ifibio_year_stats.sort_values(["Year", "N_Publications"], ascending=[True, False])
    .groupby("Year")
    .head(2)
    .copy()
)

top2_ifibio_year["Year"] = top2_ifibio_year["Year"].astype(int)

display(top2_ifibio_year)


# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(15, 7))

ax = sns.barplot(
    data=top2_ifibio_year, x="Year", y="N_Publications", hue="Author", palette="viridis"
)


# Nombre del autor arriba de cada barra
for container in ax.containers:

    for bar in container:

        if bar.get_height() > 0:

            x = bar.get_x() + bar.get_width() / 2
            y = bar.get_height()

            # Buscar año y valor correspondientes a la barra
            year_index = round(x)
            year = sorted(top2_ifibio_year["Year"].unique())[year_index]

            candidates = top2_ifibio_year[
                (top2_ifibio_year["Year"] == year)
                & (top2_ifibio_year["N_Publications"] == y)
            ]

            if len(candidates) > 0:
                name = candidates.iloc[0]["Author"]

                ax.text(
                    x, y + 0.1, name, ha="center", va="bottom", rotation=90, fontsize=8
                )


plt.xlabel("Año")
plt.ylabel("Número de publicaciones")
plt.title("Dos autores IFIBIO más prolíficos por año")

plt.xticks(rotation=45)

# Ya tenemos los nombres sobre las barras
plt.legend([], [], frameon=False)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Matriz de interacciones entre autores
# Cada celda = número de publicaciones compartidas
# ============================================================

import ast
import pandas as pd
from itertools import combinations

authors = df["Authors"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


# Lista de autores por paper
paper_authors = []

for author_list in authors:

    names = [
        author.get("AuthorClean") for author in author_list if author.get("AuthorClean")
    ]

    # Evitar duplicados dentro de un mismo paper
    names = list(dict.fromkeys(names))

    paper_authors.append(names)


# Todos los autores
all_authors = sorted({author for names in paper_authors for author in names})

print("Autores únicos:", len(all_authors))


# ============================================================
# Matriz de coautoría
# ============================================================

coauthor_matrix = pd.DataFrame(0, index=all_authors, columns=all_authors, dtype=int)


for names in paper_authors:

    for author1, author2 in combinations(names, 2):

        coauthor_matrix.loc[author1, author2] += 1
        coauthor_matrix.loc[author2, author1] += 1


display(coauthor_matrix)


# ============================================================
# Lista de interacciones
# Más práctica para NetworkX y análisis posteriores
# ============================================================

relations = coauthor_matrix.stack().rename("Weight").reset_index()

relations.columns = ["Author1", "Author2", "Weight"]

relations = (
    relations[(relations["Weight"] > 0) & (relations["Author1"] < relations["Author2"])]
    .sort_values("Weight", ascending=False)
    .reset_index(drop=True)
)

display(relations)


# ============================================================
# Guardar para recuperar rápidamente
# ============================================================

coauthor_matrix.to_pickle(os.path.join("ml_pubmed_ifibio", "coauthor_matrix.pkl"))

relations.to_pickle(os.path.join("ml_pubmed_ifibio", "coauthor_relations.pkl"))

print("Matriz:", coauthor_matrix.shape)
print("Interacciones:", len(relations))
print("Archivos guardados:")
print("  coauthor_matrix.pkl")
print("  coauthor_relations.pkl")
```

```python
# ============================================================
# Heatmap de colaboraciones
# Solo autores con más de 10 publicaciones
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

# Número de publicaciones por autor
author_n_papers = authors_df.groupby("Author")["PMID"].nunique()


# Autores con más de 10 papers
more_than_N = 5

authors_10 = author_n_papers[author_n_papers > more_than_N].sort_values(ascending=False)

print("Autores con más de 10 publicaciones:", len(authors_10))

display(authors_10.rename("N_Publications").reset_index())


# Extraer submatriz
heatmap_matrix = coauthor_matrix.loc[authors_10.index, authors_10.index]


# ============================================================
# Heatmap
# ============================================================

plt.figure(figsize=(max(10, len(authors_10) * 0.6), max(8, len(authors_10) * 0.6)))

sns.heatmap(
    heatmap_matrix,
    cmap="viridis",
    annot=True,
    fmt="d",
    linewidths=0.3,
    square=True,
    cbar_kws={"label": "Publicaciones compartidas"},
)

plt.xlabel("")
plt.ylabel("")
plt.title(
    "Colaboraciones entre autores con más de {} publicaciones".format(more_than_N)
)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Network de coautorías
#
# Nodo:
#   - Author
#   - N_Publications
#   - Affiliations
#   - IFIBIO
#
# Arista:
#   - Weight = número de papers compartidos
# ============================================================

import ast
import pandas as pd
import networkx as nx
from itertools import combinations

# ------------------------------------------------------------
# Extraer información de autores
# ------------------------------------------------------------

authors = df["Authors"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


author_records = []

for pmid, author_list in zip(df["PMID"], authors):

    for author in author_list:

        name = author.get("AuthorClean")

        if not name:
            continue

        affiliations = [
            aff.get("Affiliation")
            for aff in author.get("Affiliations", [])
            if aff.get("Affiliation")
        ]

        author_records.append(
            {"PMID": pmid, "Author": name, "Affiliations": affiliations}
        )


author_info = pd.DataFrame(author_records)


# ------------------------------------------------------------
# Número de publicaciones por autor
# ------------------------------------------------------------

author_papers = author_info.groupby("Author")["PMID"].nunique().rename("N_Publications")


# ------------------------------------------------------------
# Filiaciones únicas de cada autor
# ------------------------------------------------------------

author_affiliations = (
    author_info.explode("Affiliations")
    .dropna(subset=["Affiliations"])
    .groupby("Author")["Affiliations"]
    .agg(lambda x: sorted(set(x)))
)


# ------------------------------------------------------------
# Identificar afiliación IFIBIO
#
# Ajustar términos si ya usaste otros criterios
# ------------------------------------------------------------

author_ifibio = author_affiliations.apply(
    lambda affiliations: any(
        (
            "IFIBIO" in affiliation.upper()
            or "INSTITUTO DE FISIOLOG" in affiliation.upper()
        )
        for affiliation in affiliations
    )
).rename("IFIBIO")


# ------------------------------------------------------------
# Tabla de nodos
# ------------------------------------------------------------

nodes = pd.DataFrame(
    {
        "N_Publications": author_papers,
        "Affiliations": author_affiliations,
        "IFIBIO": author_ifibio,
    }
)

nodes["Affiliations"] = nodes["Affiliations"].apply(
    lambda x: x if isinstance(x, list) else []
)

nodes["IFIBIO"] = nodes["IFIBIO"].fillna(False)

nodes = (
    nodes.reset_index()
    .sort_values("N_Publications", ascending=False)
    .reset_index(drop=True)
)

display(nodes.head(30))
```

```python
# ============================================================
# Construcción de las aristas de coautoría
# ============================================================

paper_authors = author_info.groupby("PMID")["Author"].agg(
    lambda x: list(dict.fromkeys(x))
)


edges = {}

for names in paper_authors:

    for author1, author2 in combinations(sorted(names), 2):

        pair = (author1, author2)

        edges[pair] = edges.get(pair, 0) + 1


relations = pd.DataFrame(
    [
        {"Author1": author1, "Author2": author2, "Weight": weight}
        for (author1, author2), weight in edges.items()
    ]
)

relations = relations.sort_values("Weight", ascending=False).reset_index(drop=True)

display(relations.head(30))
```

```python
# ============================================================
# Construcción del NetworkX graph
# ============================================================

G = nx.Graph()


# Nodos
for _, row in nodes.iterrows():

    G.add_node(
        row["Author"],
        N_Publications=int(row["N_Publications"]),
        IFIBIO=bool(row["IFIBIO"]),
        Affiliations=row["Affiliations"],
    )


# Aristas
for _, row in relations.iterrows():

    G.add_edge(row["Author1"], row["Author2"], Weight=int(row["Weight"]))


print("Nodos:", G.number_of_nodes())
print("Aristas:", G.number_of_edges())
print("Componentes:", nx.number_connected_components(G))
```

```python
# ============================================================
# Agregar métricas básicas al network
# ============================================================

degree = dict(G.degree())

weighted_degree = dict(G.degree(weight="Weight"))


nx.set_node_attributes(G, degree, "Degree")

nx.set_node_attributes(G, weighted_degree, "WeightedDegree")


network_stats = pd.DataFrame(
    {
        "Author": list(G.nodes()),
        "Degree": [degree[x] for x in G.nodes()],
        "WeightedDegree": [weighted_degree[x] for x in G.nodes()],
        "N_Publications": [G.nodes[x]["N_Publications"] for x in G.nodes()],
        "IFIBIO": [G.nodes[x]["IFIBIO"] for x in G.nodes()],
        "Affiliations": [G.nodes[x]["Affiliations"] for x in G.nodes()],
    }
)


network_stats = network_stats.sort_values(
    ["WeightedDegree", "N_Publications"], ascending=False
)

display(network_stats.head(30))
```

```python
# ============================================================
# Preparar una copia compatible con GEXF
# ============================================================

G_gexf = G.copy()

for node, data in G_gexf.nodes(data=True):

    affiliations = data.get("Affiliations", [])

    if isinstance(affiliations, list):
        data["Affiliations"] = " | ".join(affiliations)


# ============================================================
# Guardar
# ============================================================

nx.write_gexf(G_gexf, os.path.join("ml_pubmed_ifibio", "coauthor_network.gexf"))

nodes.to_pickle(os.path.join("ml_pubmed_ifibio", "coauthor_nodes.pkl"))

relations.to_pickle(os.path.join("ml_pubmed_ifibio", "coauthor_relations.pkl"))

network_stats.to_pickle(os.path.join("ml_pubmed_ifibio", "coauthor_network_stats.pkl"))

print("Guardados:")
print("coauthor_network.gexf")
print("coauthor_nodes.pkl")
print("coauthor_relations.pkl")
print("coauthor_network_stats.pkl")
```

```python
# ============================================================
# Plot global del network de coautorías
#
# Tamaño del nodo  -> número de publicaciones
# Grosor de arista -> número de publicaciones compartidas
# ============================================================

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np

# Layout
pos = nx.spring_layout(G, weight="Weight", k=0.7, iterations=100, seed=42)


# Tamaño de nodos según número de publicaciones
node_sizes = [40 + G.nodes[node]["N_Publications"] * 25 for node in G.nodes()]


# Grosor de aristas según número de coautorías
edge_weights = np.array([G[u][v]["Weight"] for u, v in G.edges()])

edge_widths = 0.3 + edge_weights * 0.5


# Plot
plt.figure(figsize=(18, 18))


nx.draw_networkx_edges(G, pos, width=edge_widths, alpha=0.15)


nx.draw_networkx_nodes(G, pos, node_size=node_sizes, alpha=0.8)


plt.title("Red global de coautorías", fontsize=16)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Network global + etiquetas para autores más prolíficos
# ============================================================

more_than_N = 6

# Los labels grandes son para los autores con mas de N publicaciones
bigger_more_than_N = 20

plt.figure(figsize=(18, 18))

nx.draw_networkx_edges(G, pos, width=edge_widths, alpha=0.15)

nx.draw_networkx_nodes(G, pos, node_size=node_sizes, alpha=0.8)

# Etiquetar solamente autores con >= more_than_N papers but < bigger_more_than_N
labels_more_than_N = {
    node: node
    for node in G.nodes()
    if G.nodes[node]["N_Publications"] >= more_than_N
    and G.nodes[node]["N_Publications"] < bigger_more_than_N
}
labels_bigger_more_than_N = {
    node: node
    for node in G.nodes()
    if G.nodes[node]["N_Publications"] >= bigger_more_than_N
}


nx.draw_networkx_labels(G, pos, labels=labels_more_than_N, font_size=8)
nx.draw_networkx_labels(
    G, pos, labels=labels_bigger_more_than_N, font_size=10, font_weight="bold"
)

plt.title(
    f"Red global de coautorías (con más de {more_than_N} publicaciones)", fontsize=16
)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Network global de coautorías
#
# IFIBIO:
#   - nodos destacados
#   - conexiones IFIBIO-IFIBIO destacadas
#
# No IFIBIO:
#   - nodos grises y transparentes
#   - conexiones atenuadas
# ============================================================

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np

# Layout
pos = nx.spring_layout(G, weight="Weight", k=0.7, iterations=100, seed=42)


# ------------------------------------------------------------
# Separar nodos IFIBIO y no IFIBIO
# ------------------------------------------------------------

ifibio_nodes = [node for node in G.nodes() if G.nodes[node].get("IFIBIO", False)]

external_nodes = [node for node in G.nodes() if not G.nodes[node].get("IFIBIO", False)]


# Tamaños según número de publicaciones
ifibio_sizes = [60 + G.nodes[node]["N_Publications"] * 30 for node in ifibio_nodes]

external_sizes = [30 + G.nodes[node]["N_Publications"] * 15 for node in external_nodes]


# ------------------------------------------------------------
# Separar aristas
# ------------------------------------------------------------

ifibio_edges = []
other_edges = []

for u, v in G.edges():

    if G.nodes[u].get("IFIBIO", False) and G.nodes[v].get("IFIBIO", False):
        ifibio_edges.append((u, v))

    else:
        other_edges.append((u, v))


# Grosor de conexiones
ifibio_widths = [0.8 + G[u][v]["Weight"] * 0.6 for u, v in ifibio_edges]

other_widths = [0.2 + G[u][v]["Weight"] * 0.15 for u, v in other_edges]


# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(18, 18))


# Conexiones externas / mixtas
nx.draw_networkx_edges(
    G, pos, edgelist=other_edges, width=other_widths, edge_color="gray", alpha=0.08
)


# Conexiones IFIBIO-IFIBIO
nx.draw_networkx_edges(
    G, pos, edgelist=ifibio_edges, width=ifibio_widths, edge_color="tab:red", alpha=0.5
)


# Nodos externos
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=external_nodes,
    node_size=external_sizes,
    node_color="gray",
    alpha=0.12,
)


# Nodos IFIBIO
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=ifibio_nodes,
    node_size=ifibio_sizes,
    node_color="tab:blue",
    alpha=0.9,
)


# ------------------------------------------------------------
# Etiquetas solo para autores IFIBIO
# ------------------------------------------------------------

labels = {node: node for node in ifibio_nodes}

nx.draw_networkx_labels(G, pos, labels=labels, font_size=8)


plt.title("Red de coautorías — autores IFIBIO destacados", fontsize=16)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Network global de coautorías
#
# IFIBIO:
#   - nodos destacados
#   - conexiones IFIBIO-IFIBIO destacadas
#   - etiquetas solo para autores IFIBIO con más de la mediana
#     de publicaciones de los autores IFIBIO
# ============================================================

import matplotlib.pyplot as plt
import networkx as nx

# ------------------------------------------------------------
# Mediana de publicaciones entre autores IFIBIO
# ------------------------------------------------------------

ifibio_papers = [G.nodes[node]["N_Publications"] for node in ifibio_nodes]

median_ifibio = pd.Series(ifibio_papers).median()

print("Mediana de publicaciones de autores IFIBIO +2:", median_ifibio + 2)


# ------------------------------------------------------------
# Etiquetas solo para autores IFIBIO por encima de la mediana
# ------------------------------------------------------------

labels = {
    node: node
    for node in ifibio_nodes
    if G.nodes[node]["N_Publications"] > median_ifibio + 2
}


# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(18, 18))


# Conexiones externas / mixtas
nx.draw_networkx_edges(
    G, pos, edgelist=other_edges, width=other_widths, edge_color="gray", alpha=0.08
)


# Conexiones IFIBIO-IFIBIO
nx.draw_networkx_edges(
    G, pos, edgelist=ifibio_edges, width=ifibio_widths, edge_color="tab:red", alpha=0.50
)


# Nodos externos
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=external_nodes,
    node_size=external_sizes,
    node_color="gray",
    alpha=0.12,
)


# Nodos IFIBIO
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=ifibio_nodes,
    node_size=ifibio_sizes,
    node_color="tab:blue",
    alpha=0.9,
)


# Etiquetas seleccionadas
nx.draw_networkx_labels(G, pos, labels=labels, font_size=8)


plt.title(
    f"Red de coautorías — autores IFIBIO destacados ({median_ifibio + 2} publicaciones)",
    fontsize=16,
)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Cálculo de clusters de coautoría con Louvain
# ============================================================

import networkx as nx
import pandas as pd

# Detectar comunidades
communities = nx.community.louvain_communities(
    G, weight="Weight", resolution=1.0, seed=42
)


print("Número de clusters:", len(communities))


# ============================================================
# Asignar cluster a cada autor
# ============================================================

cluster_map = {}

for cluster_id, community in enumerate(communities):

    for author in community:
        cluster_map[author] = cluster_id


nx.set_node_attributes(G, cluster_map, "Cluster")


# ============================================================
# Tabla de autores y clusters
# ============================================================

cluster_df = pd.DataFrame(
    {
        "Author": list(G.nodes()),
        "Cluster": [G.nodes[node]["Cluster"] for node in G.nodes()],
        "N_Publications": [G.nodes[node]["N_Publications"] for node in G.nodes()],
        "IFIBIO": [G.nodes[node].get("IFIBIO", False) for node in G.nodes()],
    }
)


cluster_df = cluster_df.sort_values(
    ["Cluster", "N_Publications"], ascending=[True, False]
).reset_index(drop=True)


display(cluster_df)
```

```python
# ============================================================
# Estadísticas por cluster
# ============================================================

cluster_stats = (
    cluster_df.groupby("Cluster")
    .agg(
        N_Authors=("Author", "count"),
        N_IFIBIO=("IFIBIO", "sum"),
        Total_Publications=("N_Publications", "sum"),
        Mean_Publications=("N_Publications", "mean"),
        Median_Publications=("N_Publications", "median"),
    )
    .sort_values("N_Authors", ascending=False)
    .reset_index()
)


display(cluster_stats)
```

```python
# ============================================================
# Modularidad de la partición
# ============================================================

modularity = nx.community.modularity(G, communities, weight="Weight")

print("Modularidad:", round(modularity, 4))
```

```python
# ============================================================
# Guardar clusters
# ============================================================

cluster_df.to_pickle(os.path.join("ml_pubmed_ifibio", "coauthor_clusters.pkl"))

cluster_stats.to_pickle(os.path.join("ml_pubmed_ifibio", "coauthor_cluster_stats.pkl"))

print("Guardado:")
print("coauthor_clusters.pkl")
print("coauthor_cluster_stats.pkl")
```

```python
# ============================================================
# Plot del network por clusters
# Solo clusters con:
#   - más de 2 publicaciones
#   - más de 4 integrantes
# ============================================================

import matplotlib.pyplot as plt
import networkx as nx
import seaborn as sns

# ------------------------------------------------------------
# Publicaciones únicas por cluster
# ------------------------------------------------------------

author_cluster = cluster_df[["Author", "Cluster"]]

cluster_papers = (
    author_info.merge(author_cluster, on="Author", how="left")
    .groupby("Cluster")["PMID"]
    .nunique()
    .rename("N_Publications")
)


# ------------------------------------------------------------
# Número de integrantes por cluster
# ------------------------------------------------------------

cluster_members = cluster_df.groupby("Cluster")["Author"].nunique().rename("N_Authors")


cluster_filter = pd.concat([cluster_papers, cluster_members], axis=1)


# Clusters que cumplen los criterios
valid_clusters = cluster_filter[
    (cluster_filter["N_Publications"] > 2) & (cluster_filter["N_Authors"] > 4)
].index


display(
    cluster_filter.loc[valid_clusters].sort_values("N_Publications", ascending=False)
)
```

```python
# ============================================================
# Autores centrales de cada cluster
#
# Métricas:
#   Degree            = número de colaboradores directos
#   WeightedDegree    = intensidad total de colaboraciones
#   Betweenness       = rol de puente dentro del cluster
#   PageRank          = centralidad considerando peso de vínculos
# ============================================================

import pandas as pd
import networkx as nx

centrality_records = []


for cluster in sorted(valid_clusters):

    cluster_nodes = [
        node for node in G.nodes() if G.nodes[node].get("Cluster") == cluster
    ]

    Gc = G.subgraph(cluster_nodes).copy()

    # Degree
    degree = dict(Gc.degree())

    # Weighted degree
    weighted_degree = dict(Gc.degree(weight="Weight"))

    # Para betweenness, mayor Weight = vínculo más fuerte,
    # por lo tanto usamos distancia = 1 / Weight
    for u, v in Gc.edges():
        Gc[u][v]["Distance"] = 1 / Gc[u][v]["Weight"]

    betweenness = nx.betweenness_centrality(Gc, weight="Distance", normalized=True)

    pagerank = nx.pagerank(Gc, weight="Weight")

    for node in Gc.nodes():

        centrality_records.append(
            {
                "Cluster": cluster,
                "Author": node,
                "N_Publications": G.nodes[node]["N_Publications"],
                "IFIBIO": G.nodes[node].get("IFIBIO", False),
                "Degree": degree[node],
                "WeightedDegree": weighted_degree[node],
                "Betweenness": betweenness[node],
                "PageRank": pagerank[node],
            }
        )


cluster_centrality = pd.DataFrame(centrality_records)


# ============================================================
# Ranking combinado de centralidad
# ============================================================

cluster_centrality["CentralityScore"] = (
    cluster_centrality.groupby("Cluster")[["WeightedDegree", "Betweenness", "PageRank"]]
    .transform(lambda x: (x - x.min()) / (x.max() - x.min()))
    .fillna(0)
    .mean(axis=1)
)


cluster_centrality = cluster_centrality.sort_values(
    ["Cluster", "CentralityScore"], ascending=[True, False]
).reset_index(drop=True)


display(cluster_centrality)
```

```python
# ============================================================
# Autor más central de cada cluster
# ============================================================

central_author_cluster = (
    cluster_centrality.sort_values(
        ["Cluster", "CentralityScore"], ascending=[True, False]
    )
    .groupby("Cluster")
    .head(1)
    .reset_index(drop=True)
)


display(
    central_author_cluster[
        [
            "Cluster",
            "Author",
            "IFIBIO",
            "N_Publications",
            "Degree",
            "WeightedDegree",
            "Betweenness",
            "PageRank",
            "CentralityScore",
        ]
    ]
)
```

```python
# ============================================================
# Análisis de Keywords
# ============================================================

import ast
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# ------------------------------------------------------------
# Parsear la columna Keywords
# ------------------------------------------------------------

keywords = df["Keywords"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


# Ver estructura de algunos registros
display(keywords[keywords.apply(len) > 0].head(10))

# ============================================================
# Extraer todas las keywords
# ============================================================

all_keywords = []

for keyword_list in keywords:

    for keyword in keyword_list:

        # Si la keyword está almacenada como diccionario
        if isinstance(keyword, dict):

            term = keyword.get("Keyword") or keyword.get("Text") or keyword.get("Term")

        # Si está almacenada directamente como string
        else:
            term = keyword

        if term:
            all_keywords.append(str(term).strip())


print("Keywords totales:", len(all_keywords))
print("Keywords únicas:", len(set(all_keywords)))

# ============================================================
# Conteo de keywords
# ============================================================

keyword_counts = (
    pd.Series(all_keywords)
    .value_counts()
    .rename_axis("Keyword")
    .reset_index(name="Count")
)

keyword_counts["Percent"] = (
    keyword_counts["Count"] / keyword_counts["Count"].sum() * 100
).round(2)

display(keyword_counts)

# ============================================================
# Estadísticas generales
# ============================================================

papers_with_keywords = keywords.apply(len) > 0

print("Número total de papers:", len(df))
print("Papers con keywords:", papers_with_keywords.sum())
print("Papers sin keywords:", (~papers_with_keywords).sum())

print("Papers con keywords (%):", round(papers_with_keywords.mean() * 100, 2))

print()
print("Keywords totales:", len(all_keywords))
print("Keywords únicas:", len(set(all_keywords)))

print("Promedio de keywords por paper:", round(keywords.apply(len).mean(), 2))

print("Mediana de keywords por paper:", keywords.apply(len).median())

print("Máximo de keywords en un paper:", keywords.apply(len).max())

# ============================================================
# Top 30 keywords
# ============================================================

top_keywords = keyword_counts.head(30)

plt.figure(figsize=(10, 10))

sns.barplot(
    data=top_keywords.sort_values("Count"),
    x="Count",
    y="Keyword",
    hue="Count",
    palette="viridis",
    legend=False,
)

plt.xlabel("Número de apariciones")
plt.ylabel("")
plt.title("30 keywords más frecuentes")

plt.tight_layout()
plt.show()

# ============================================================
# Distribución del número de keywords por paper
# ============================================================

df["N_Keywords"] = keywords.apply(len)

plt.figure(figsize=(9, 5))

sns.histplot(df["N_Keywords"], discrete=True)

plt.xlabel("Número de keywords")
plt.ylabel("Número de publicaciones")
plt.title("Distribución de keywords por publicación")

plt.tight_layout()
plt.show()

display(keyword_counts[["Keyword", "Count", "Percent"]])
```

```python
# ============================================================
# Network de co-ocurrencia de keywords
#
# Nodo:
#   - Keyword
#   - N_Papers = cantidad de papers donde aparece
#
# Arista:
#   - Weight = cantidad de papers donde ambas keywords aparecen juntas
# ============================================================

import pandas as pd
import networkx as nx
from itertools import combinations
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# Limpiar keywords por paper
# ------------------------------------------------------------

paper_keywords = []

for keyword_list in keywords:

    clean_keywords = []

    for keyword in keyword_list:

        if isinstance(keyword, dict):
            term = keyword.get("Keyword") or keyword.get("Text") or keyword.get("Term")
        else:
            term = keyword

        if term:
            clean_keywords.append(str(term).strip())

    # evitar duplicados dentro del mismo paper
    clean_keywords = list(dict.fromkeys(clean_keywords))

    paper_keywords.append(clean_keywords)


# ------------------------------------------------------------
# Frecuencia de cada keyword
# ------------------------------------------------------------

keyword_frequency = {}

for kw_list in paper_keywords:
    for kw in kw_list:
        keyword_frequency[kw] = keyword_frequency.get(kw, 0) + 1


# ------------------------------------------------------------
# Co-ocurrencias
# ------------------------------------------------------------

keyword_edges = {}

for kw_list in paper_keywords:

    for kw1, kw2 in combinations(sorted(kw_list), 2):

        pair = (kw1, kw2)

        keyword_edges[pair] = keyword_edges.get(pair, 0) + 1


# ------------------------------------------------------------
# Crear graph
# ------------------------------------------------------------

G_keywords = nx.Graph()


# Nodos
for keyword, n_papers in keyword_frequency.items():

    G_keywords.add_node(keyword, N_Papers=n_papers)


# Aristas
for (kw1, kw2), weight in keyword_edges.items():

    G_keywords.add_edge(kw1, kw2, Weight=weight)


print("Keywords:", G_keywords.number_of_nodes())
print("Relaciones:", G_keywords.number_of_edges())
```

```python
# ============================================================
# Tablas de nodos y relaciones
# ============================================================

keyword_nodes = (
    pd.DataFrame(
        [
            {"Keyword": node, "N_Papers": data["N_Papers"]}
            for node, data in G_keywords.nodes(data=True)
        ]
    )
    .sort_values("N_Papers", ascending=False)
    .reset_index(drop=True)
)


keyword_relations = (
    pd.DataFrame(
        [
            {"Keyword1": u, "Keyword2": v, "Weight": data["Weight"]}
            for u, v, data in G_keywords.edges(data=True)
        ]
    )
    .sort_values("Weight", ascending=False)
    .reset_index(drop=True)
)


display(keyword_nodes)
display(keyword_relations)
```

```python
# ============================================================
# Análisis de términos MeSH
# ============================================================

import ast
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# Parsear columna MeSH
# ------------------------------------------------------------

mesh = df["MeSH"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


# Ver estructura
display(mesh[mesh.apply(len) > 0].head(10))


# ============================================================
# Extraer todos los términos MeSH
# ============================================================

all_mesh = []

for mesh_list in mesh:

    for item in mesh_list:

        if isinstance(item, dict):

            term = (
                item.get("DescriptorName")
                or item.get("Descriptor")
                or item.get("MeSH")
                or item.get("Term")
                or item.get("Name")
            )

        else:
            term = item

        if term:
            all_mesh.append(str(term).strip())


print("Términos MeSH totales:", len(all_mesh))
print("Términos MeSH únicos:", len(set(all_mesh)))

# ============================================================
# Conteo de términos MeSH
# ============================================================

mesh_counts = (
    pd.Series(all_mesh).value_counts().rename_axis("MeSH").reset_index(name="Count")
)

mesh_counts["Percent"] = (
    mesh_counts["Count"] / mesh_counts["Count"].sum() * 100
).round(2)


display(mesh_counts)

# ============================================================
# Conteo de términos MeSH
# ============================================================

mesh_counts = (
    pd.Series(all_mesh).value_counts().rename_axis("MeSH").reset_index(name="Count")
)

mesh_counts["Percent"] = (
    mesh_counts["Count"] / mesh_counts["Count"].sum() * 100
).round(2)


display(mesh_counts)

# ============================================================
# Estadísticas generales de MeSH
# ============================================================

papers_with_mesh = mesh.apply(len) > 0

df["N_MeSH"] = mesh.apply(len)


print("Número total de papers:", len(df))

print("Papers con MeSH:", papers_with_mesh.sum())

print("Papers sin MeSH:", (~papers_with_mesh).sum())

print("Papers con MeSH (%):", round(papers_with_mesh.mean() * 100, 2))

print()

print("Términos MeSH totales:", len(all_mesh))

print("Términos MeSH únicos:", len(set(all_mesh)))

print("Promedio de términos MeSH por paper:", round(df["N_MeSH"].mean(), 2))

print("Mediana de términos MeSH por paper:", df["N_MeSH"].median())

print("Máximo de términos MeSH en un paper:", df["N_MeSH"].max())

# ============================================================
# Top 30 términos MeSH
# ============================================================

top_mesh = mesh_counts.head(30)

plt.figure(figsize=(10, 10))

sns.barplot(
    data=top_mesh.sort_values("Count"),
    x="Count",
    y="MeSH",
    hue="Count",
    palette="viridis",
    legend=False,
)

plt.xlabel("Número de apariciones")
plt.ylabel("")
plt.title("30 términos MeSH más frecuentes")

plt.tight_layout()
plt.show()

# ============================================================
# Distribución del número de términos MeSH por paper
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(df["N_MeSH"], discrete=True)

plt.xlabel("Número de términos MeSH")
plt.ylabel("Número de publicaciones")
plt.title("Distribución de términos MeSH por publicación")

plt.tight_layout()
plt.show()

# ============================================================
# Preparar términos MeSH por paper para network
# ============================================================

paper_mesh = []

for mesh_list in mesh:

    clean_mesh = []

    for item in mesh_list:

        if isinstance(item, dict):

            term = (
                item.get("DescriptorName")
                or item.get("Descriptor")
                or item.get("MeSH")
                or item.get("Term")
                or item.get("Name")
            )

        else:
            term = item

        if term:
            clean_mesh.append(str(term).strip())

    # Evitar duplicados dentro de un paper
    clean_mesh = list(dict.fromkeys(clean_mesh))

    paper_mesh.append(clean_mesh)

# ============================================================
# Network de co-ocurrencia MeSH
#
# Nodo:
#   N_Papers = papers donde aparece el término
#
# Arista:
#   Weight = papers donde ambos términos aparecen juntos
# ============================================================

import networkx as nx
from itertools import combinations

# Frecuencia de términos
mesh_frequency = {}

for mesh_list in paper_mesh:

    for term in mesh_list:

        mesh_frequency[term] = mesh_frequency.get(term, 0) + 1


# Co-ocurrencias
mesh_edges = {}

for mesh_list in paper_mesh:

    for mesh1, mesh2 in combinations(sorted(mesh_list), 2):

        pair = (mesh1, mesh2)

        mesh_edges[pair] = mesh_edges.get(pair, 0) + 1


# ------------------------------------------------------------
# Crear grafo
# ------------------------------------------------------------

G_mesh = nx.Graph()


# Nodos
for term, n_papers in mesh_frequency.items():

    G_mesh.add_node(term, N_Papers=n_papers)


# Aristas
for (mesh1, mesh2), weight in mesh_edges.items():

    G_mesh.add_edge(mesh1, mesh2, Weight=weight)


print("Términos MeSH:", G_mesh.number_of_nodes())

print("Relaciones:", G_mesh.number_of_edges())


# ============================================================
# Tablas de nodos y relaciones
# ============================================================

mesh_nodes = (
    pd.DataFrame(
        [
            {"MeSH": node, "N_Papers": data["N_Papers"]}
            for node, data in G_mesh.nodes(data=True)
        ]
    )
    .sort_values("N_Papers", ascending=False)
    .reset_index(drop=True)
)


mesh_relations = (
    pd.DataFrame(
        [
            {"MeSH1": u, "MeSH2": v, "Weight": data["Weight"]}
            for u, v, data in G_mesh.edges(data=True)
        ]
    )
    .sort_values("Weight", ascending=False)
    .reset_index(drop=True)
)


display(mesh_nodes)
display(mesh_relations)

# ============================================================
# Guardar network MeSH
# ============================================================

nx.write_gexf(G_mesh, os.path.join("ml_pubmed_ifibio", "mesh_network.gexf"))

mesh_nodes.to_pickle(os.path.join("ml_pubmed_ifibio", "mesh_nodes.pkl"))

mesh_relations.to_pickle(os.path.join("ml_pubmed_ifibio", "mesh_relations.pkl"))

print("Guardados:")
print("mesh_network.gexf")
print("mesh_nodes.pkl")
print("mesh_relations.pkl")

# ============================================================
# Plot global del network MeSH
#
# Mostrar términos presentes en >= 3 papers
# ============================================================

selected_mesh = [node for node in G_mesh.nodes() if G_mesh.nodes[node]["N_Papers"] >= 3]


Gm = G_mesh.subgraph(selected_mesh).copy()


print("Términos graficados:", Gm.number_of_nodes())

print("Relaciones graficadas:", Gm.number_of_edges())


# ------------------------------------------------------------
# Layout
# ------------------------------------------------------------

pos_mesh = nx.spring_layout(Gm, weight="Weight", k=0.9, iterations=150, seed=42)


# ------------------------------------------------------------
# Tamaño de nodos
# ------------------------------------------------------------

node_sizes = [80 + Gm.nodes[node]["N_Papers"] * 50 for node in Gm.nodes()]


# ------------------------------------------------------------
# Grosor de aristas
# ------------------------------------------------------------

edge_widths = [0.2 + Gm[u][v]["Weight"] * 0.5 for u, v in Gm.edges()]


# ------------------------------------------------------------
# Color según frecuencia
# ------------------------------------------------------------

node_colors = [Gm.nodes[node]["N_Papers"] for node in Gm.nodes()]


# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(18, 18))


nx.draw_networkx_edges(Gm, pos_mesh, width=edge_widths, edge_color="gray", alpha=0.15)


nodes_plot = nx.draw_networkx_nodes(
    Gm,
    pos_mesh,
    node_size=node_sizes,
    node_color=node_colors,
    cmap="viridis",
    alpha=0.9,
)


# Etiquetas solo para términos por encima de la mediana
median_frequency = pd.Series(
    [Gm.nodes[node]["N_Papers"] for node in Gm.nodes()]
).median()


labels = {
    node: node for node in Gm.nodes() if Gm.nodes[node]["N_Papers"] > median_frequency
}


nx.draw_networkx_labels(Gm, pos_mesh, labels=labels, font_size=8)


plt.colorbar(nodes_plot, label="Número de publicaciones", shrink=0.6)

plt.title("Network de co-ocurrencia de términos MeSH", fontsize=16)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Gran network integrado:
#   - Autores
#   - Términos MeSH
#   - Keywords
#
# Tipos de relaciones:
#   Author-Author   -> coautoría
#   MeSH-MeSH       -> co-ocurrencia
#   Keyword-Keyword -> co-ocurrencia
#   Author-MeSH     -> autor publicó paper con ese MeSH
#   Author-Keyword  -> autor publicó paper con esa keyword
#   MeSH-Keyword    -> aparecen juntos en el mismo paper
# ============================================================

import ast
import pandas as pd
import networkx as nx
from itertools import combinations

# ------------------------------------------------------------
# Preparar datos por paper
# ------------------------------------------------------------

authors_parsed = df["Authors"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)

mesh_parsed = df["MeSH"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)

keywords_parsed = df["Keywords"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


paper_data = []


for pmid, author_list, mesh_list, keyword_list in zip(
    df["PMID"], authors_parsed, mesh_parsed, keywords_parsed
):

    # -------------------------
    # Autores
    # -------------------------

    paper_authors = [
        author.get("AuthorClean") for author in author_list if author.get("AuthorClean")
    ]

    paper_authors = list(dict.fromkeys(paper_authors))

    # -------------------------
    # MeSH
    # -------------------------

    paper_mesh = []

    for item in mesh_list:

        if isinstance(item, dict):

            term = (
                item.get("DescriptorName")
                or item.get("Descriptor")
                or item.get("MeSH")
                or item.get("Term")
                or item.get("Name")
            )

        else:
            term = item

        if term:
            paper_mesh.append(str(term).strip())

    paper_mesh = list(dict.fromkeys(paper_mesh))

    # -------------------------
    # Keywords
    # -------------------------

    paper_keywords = []

    for item in keyword_list:

        if isinstance(item, dict):

            term = item.get("Keyword") or item.get("Text") or item.get("Term")

        else:
            term = item

        if term:
            paper_keywords.append(str(term).strip())

    paper_keywords = list(dict.fromkeys(paper_keywords))

    paper_data.append(
        {
            "PMID": pmid,
            "Authors": paper_authors,
            "MeSH": paper_mesh,
            "Keywords": paper_keywords,
        }
    )
```

```python
# ============================================================
# Crear grafo integrado
# ============================================================

G_all = nx.Graph()


# ------------------------------------------------------------
# Agregar nodos
# Usamos prefijos para evitar colisiones de nombres
# ------------------------------------------------------------

for paper in paper_data:

    for author in paper["Authors"]:

        node = "AUTHOR::" + author

        if not G_all.has_node(node):
            G_all.add_node(node, Label=author, Type="Author", N_Papers=0)

        G_all.nodes[node]["N_Papers"] += 1

    for mesh_term in paper["MeSH"]:

        node = "MESH::" + mesh_term

        if not G_all.has_node(node):
            G_all.add_node(node, Label=mesh_term, Type="MeSH", N_Papers=0)

        G_all.nodes[node]["N_Papers"] += 1

    for keyword in paper["Keywords"]:

        node = "KEYWORD::" + keyword

        if not G_all.has_node(node):
            G_all.add_node(node, Label=keyword, Type="Keyword", N_Papers=0)

        G_all.nodes[node]["N_Papers"] += 1

# ============================================================
# Agregar aristas
# ============================================================

for paper in paper_data:

    authors_nodes = ["AUTHOR::" + x for x in paper["Authors"]]

    mesh_nodes_paper = ["MESH::" + x for x in paper["MeSH"]]

    keyword_nodes_paper = ["KEYWORD::" + x for x in paper["Keywords"]]

    # --------------------------------------------------------
    # Author - Author
    # --------------------------------------------------------

    for u, v in combinations(sorted(authors_nodes), 2):

        if G_all.has_edge(u, v):
            G_all[u][v]["Weight"] += 1

        else:
            G_all.add_edge(u, v, Weight=1, Relation="Author-Author")

    # --------------------------------------------------------
    # MeSH - MeSH
    # --------------------------------------------------------

    for u, v in combinations(sorted(mesh_nodes_paper), 2):

        if G_all.has_edge(u, v):
            G_all[u][v]["Weight"] += 1

        else:
            G_all.add_edge(u, v, Weight=1, Relation="MeSH-MeSH")

    # --------------------------------------------------------
    # Keyword - Keyword
    # --------------------------------------------------------

    for u, v in combinations(sorted(keyword_nodes_paper), 2):

        if G_all.has_edge(u, v):
            G_all[u][v]["Weight"] += 1

        else:
            G_all.add_edge(u, v, Weight=1, Relation="Keyword-Keyword")

    # --------------------------------------------------------
    # Author - MeSH
    # --------------------------------------------------------

    for author in authors_nodes:

        for mesh_term in mesh_nodes_paper:

            if G_all.has_edge(author, mesh_term):
                G_all[author][mesh_term]["Weight"] += 1

            else:
                G_all.add_edge(author, mesh_term, Weight=1, Relation="Author-MeSH")

    # --------------------------------------------------------
    # Author - Keyword
    # --------------------------------------------------------

    for author in authors_nodes:

        for keyword in keyword_nodes_paper:

            if G_all.has_edge(author, keyword):
                G_all[author][keyword]["Weight"] += 1

            else:
                G_all.add_edge(author, keyword, Weight=1, Relation="Author-Keyword")

    # --------------------------------------------------------
    # MeSH - Keyword
    # --------------------------------------------------------

    for mesh_term in mesh_nodes_paper:

        for keyword in keyword_nodes_paper:

            if G_all.has_edge(mesh_term, keyword):
                G_all[mesh_term][keyword]["Weight"] += 1

            else:
                G_all.add_edge(mesh_term, keyword, Weight=1, Relation="MeSH-Keyword")

# ============================================================
# Incorporar afiliación IFIBIO a los nodos de autores
# ============================================================

for node in G_all.nodes():

    if G_all.nodes[node]["Type"] != "Author":
        continue

    author_name = G_all.nodes[node]["Label"]

    if author_name in G.nodes():

        G_all.nodes[node]["IFIBIO"] = bool(G.nodes[author_name].get("IFIBIO", False))

        G_all.nodes[node]["Affiliations"] = " | ".join(
            G.nodes[author_name].get("Affiliations", [])
        )

    else:
        G_all.nodes[node]["IFIBIO"] = False
        G_all.nodes[node]["Affiliations"] = ""

# ============================================================
# Estadísticas generales
# ============================================================

print("Nodos totales:", G_all.number_of_nodes())
print("Aristas totales:", G_all.number_of_edges())

print()

print("Autores:", sum(G_all.nodes[n]["Type"] == "Author" for n in G_all.nodes()))

print("MeSH:", sum(G_all.nodes[n]["Type"] == "MeSH" for n in G_all.nodes()))

print("Keywords:", sum(G_all.nodes[n]["Type"] == "Keyword" for n in G_all.nodes()))


# ============================================================
# Tabla de nodos
# ============================================================

all_nodes = pd.DataFrame(
    [
        {
            "Node": node,
            "Label": data.get("Label"),
            "Type": data.get("Type"),
            "N_Papers": data.get("N_Papers"),
            "IFIBIO": data.get("IFIBIO", False),
            "Affiliations": data.get("Affiliations", ""),
        }
        for node, data in G_all.nodes(data=True)
    ]
)

display(all_nodes.sort_values("N_Papers", ascending=False))

# ============================================================
# Tabla de relaciones
# ============================================================

all_edges = pd.DataFrame(
    [
        {"Node1": u, "Node2": v, "Weight": data["Weight"], "Relation": data["Relation"]}
        for u, v, data in G_all.edges(data=True)
    ]
)

display(all_edges.sort_values("Weight", ascending=False))

# ============================================================
# Guardar
# ============================================================

nx.write_gexf(
    G_all,
    os.path.join("ml_pubmed_ifibio", "integrated_author_mesh_keyword_network.gexf"),
)

all_nodes.to_pickle(os.path.join("ml_pubmed_ifibio", "integrated_network_nodes.pkl"))

all_edges.to_pickle(os.path.join("ml_pubmed_ifibio", "integrated_network_edges.pkl"))

nx.write_gpickle = None
```

```python
# ============================================================
# A. Clustering temático de autores
#
# Author × (MeSH + Keywords)
# -> matriz de frecuencias
# -> TF-IDF
# -> TruncatedSVD
# -> KMeans
# ============================================================

import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.decomposition import TruncatedSVD
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# ============================================================
# Construir registros Autor - término
# usando paper_data creado anteriormente
# ============================================================

records = []

for paper in paper_data:

    terms = ["MESH::" + x for x in paper["MeSH"]] + [
        "KEYWORD::" + x for x in paper["Keywords"]
    ]

    for author in paper["Authors"]:

        for term in terms:

            records.append({"Author": author, "Term": term})


author_term_df = pd.DataFrame(records)

display(author_term_df.head())
print("Registros:", len(author_term_df))


# ============================================================
# Matriz Autor × término
#
# Valor = cantidad de papers que relacionan
# al autor con ese término
# ============================================================

author_term_matrix = pd.crosstab(author_term_df["Author"], author_term_df["Term"])

print("Autores:", author_term_matrix.shape[0])
print("Términos:", author_term_matrix.shape[1])

display(author_term_matrix)

# ============================================================
# Filtrar autores con muy pocas publicaciones
#
# Mantener autores con >= 3 publicaciones
# ============================================================

author_publications = author_info.groupby("Author")["PMID"].nunique()

selected_authors = author_publications[author_publications >= 3].index

author_term_matrix_filtered = author_term_matrix.loc[
    author_term_matrix.index.intersection(selected_authors)
]

print("Autores usados:", author_term_matrix_filtered.shape[0])

print("Términos:", author_term_matrix_filtered.shape[1])

# ============================================================
# TF-IDF
#
# Reduce el peso de términos demasiado generales
# y aumenta el de términos más específicos
# ============================================================

tfidf = TfidfTransformer()

X_tfidf = tfidf.fit_transform(author_term_matrix_filtered)

print("Forma TF-IDF:", X_tfidf.shape)


# ============================================================
# Reducir dimensionalidad con TruncatedSVD
# ============================================================

n_components = min(20, X_tfidf.shape[0] - 1, X_tfidf.shape[1] - 1)

svd = TruncatedSVD(n_components=n_components, random_state=42)

X_svd = svd.fit_transform(X_tfidf)

print("Dimensiones originales:", X_tfidf.shape[1])
print("Dimensiones reducidas:", X_svd.shape[1])

print("Varianza explicada:", round(svd.explained_variance_ratio_.sum() * 100, 2), "%")
```

```python
# ============================================================
# Elegir número de clusters con método del codo
# ============================================================

k_values = range(2, 30)

inertias = []

for k in k_values:

    model = KMeans(n_clusters=k, random_state=42, n_init=20)

    model.fit(X_svd)

    inertias.append(model.inertia_)


plt.figure(figsize=(8, 5))

plt.plot(list(k_values), inertias, marker="o")

plt.xlabel("Número de clusters")
plt.ylabel("Inercia")
plt.title("Método del codo")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Silhouette score para comparar número de clusters
# ============================================================

from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in k_values:

    model = KMeans(n_clusters=k, random_state=42, n_init=20)

    labels = model.fit_predict(X_svd)

    score = silhouette_score(X_svd, labels)

    silhouette_scores.append(score)


silhouette_df = pd.DataFrame({"K": list(k_values), "Silhouette": silhouette_scores})

display(silhouette_df)


plt.figure(figsize=(8, 5))

plt.plot(silhouette_df["K"], silhouette_df["Silhouette"], marker="o")

plt.xlabel("Número de clusters")
plt.ylabel("Silhouette score")
plt.title("Calidad del clustering")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Seleccionar automáticamente el K con mejor silhouette
# ============================================================

best_k = int(silhouette_df.loc[silhouette_df["Silhouette"].idxmax(), "K"])

print("Número de clusters seleccionado:", best_k)
```

```python
# ============================================================
# KMeans final
# ============================================================

kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=50)

clusters = kmeans.fit_predict(X_svd)


author_clusters = pd.DataFrame(
    {"Author": author_term_matrix_filtered.index, "Cluster": clusters}
)


author_clusters = author_clusters.merge(
    author_publications.rename("N_Publications"),
    left_on="Author",
    right_index=True,
    how="left",
)


author_clusters = author_clusters.sort_values(
    ["Cluster", "N_Publications"], ascending=[True, False]
).reset_index(drop=True)

display(author_clusters)


# ============================================================
# Número de autores por cluster
# ============================================================

cluster_sizes = (
    author_clusters.groupby("Cluster")
    .agg(
        N_Authors=("Author", "count"),
        Total_Publications=("N_Publications", "sum"),
        Mean_Publications=("N_Publications", "mean"),
    )
    .reset_index()
)

display(cluster_sizes)


# ============================================================
# Visualización 2D con SVD
# ============================================================

svd_2d = TruncatedSVD(n_components=2, random_state=42)

X_2d = svd_2d.fit_transform(X_tfidf)


plot_df = pd.DataFrame(
    {
        "X": X_2d[:, 0],
        "Y": X_2d[:, 1],
        "Author": author_term_matrix_filtered.index,
        "Cluster": clusters,
        "N_Publications": [
            author_publications.get(author, 0)
            for author in author_term_matrix_filtered.index
        ],
    }
)


plt.figure(figsize=(12, 9))

scatter = plt.scatter(
    plot_df["X"],
    plot_df["Y"],
    c=plot_df["Cluster"],
    s=50 + plot_df["N_Publications"] * 15,
    cmap="tab10",
    alpha=0.8,
)


# Etiquetar autores más productivos
threshold = plot_df["N_Publications"].quantile(0.80)

for _, row in plot_df.iterrows():

    if row["N_Publications"] >= threshold:

        plt.annotate(row["Author"], (row["X"], row["Y"]), fontsize=8)


plt.xlabel("SVD 1")
plt.ylabel("SVD 2")
plt.title("Clusters temáticos de autores")

plt.colorbar(scatter, label="Cluster")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Obtener términos característicos de cada cluster
# ============================================================

feature_names = np.array(author_term_matrix_filtered.columns)

cluster_terms = []


for cluster in range(best_k):

    cluster_mask = clusters == cluster

    mean_tfidf = np.asarray(X_tfidf[cluster_mask].mean(axis=0)).ravel()

    top_indices = mean_tfidf.argsort()[::-1][:15]

    for rank, idx in enumerate(top_indices, start=1):

        cluster_terms.append(
            {
                "Cluster": cluster,
                "Rank": rank,
                "Term": feature_names[idx],
                "Mean_TFIDF": mean_tfidf[idx],
            }
        )


cluster_terms_df = pd.DataFrame(cluster_terms)

display(cluster_terms_df)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

# Descargar recursos necesarios de NLTK (solo la primera vez)
nltk.download("punkt")
nltk.download("stopwords")

# Suponiendo que tienes un DataFrame llamado paper_df con la columna 'Abstract'
# Si tu DataFrame se llama diferente, ajusta el nombre


# Preprocesamiento básico de los abstracts
def preprocess_text(text):
    # Convertir a minúsculas
    text = text.lower()
    # Tokenizar
    tokens = word_tokenize(text)
    # Eliminar puntuación y stopwords
    stop_words = set(stopwords.words("english"))
    tokens = [word for word in tokens if word.isalpha() and word not in stop_words]
    return tokens


# Aplicar el preprocesamiento a la columna 'Abstract'
df["Tokens"] = df["Abstract"].fillna("").apply(preprocess_text)

# Mostrar algunos ejemplos
display(df[["Abstract", "Tokens"]].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# ============================================================
# TF-IDF sobre los abstracts preprocesados
# ============================================================

# Unir los tokens en texto para cada abstract
df["Clean_Abstract"] = df["Tokens"].apply(lambda tokens: " ".join(tokens))

# Vectorización TF-IDF
vectorizer = TfidfVectorizer(max_features=1000, stop_words="english")
X_tfidf_abs = vectorizer.fit_transform(df["Clean_Abstract"])
feature_names = vectorizer.get_feature_names_out()

# Sumar TF-IDF por término para obtener importancia global
tfidf_sums = X_tfidf_abs.sum(axis=0).A1
tfidf_terms = pd.DataFrame({"Term": feature_names, "TFIDF": tfidf_sums}).sort_values(
    "TFIDF", ascending=False
)

# ============================================================
# Plot: Top términos por TF-IDF
# ============================================================

top_n = 30
plt.figure(figsize=(10, 6))
plt.barh(
    tfidf_terms.head(top_n).iloc[::-1]["Term"],
    tfidf_terms.head(top_n).iloc[::-1]["TFIDF"],
    color="tab:blue",
)
plt.xlabel("TF-IDF total")
plt.title(f"Top {top_n} términos por TF-IDF en abstracts")
plt.tight_layout()
plt.show()

# ============================================================
# Wordcloud de términos más relevantes
# ============================================================

wordcloud = WordCloud(
    width=1200, height=600, background_color="white", colormap="tab10"
).generate_from_frequencies(dict(zip(tfidf_terms["Term"], tfidf_terms["TFIDF"])))

plt.figure(figsize=(14, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Wordcloud de términos relevantes (TF-IDF)")
plt.show()

# ============================================================
# Mostrar tabla de términos más relevantes
# ============================================================

display(tfidf_terms.head(50))
```

```python
# ============================================================
# 11. KMeans sobre los abstracts
# Comparación de número de clusters
# ============================================================

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

svd_model = TruncatedSVD(n_components=50, random_state=42)

X_abs_svd = svd_model.fit_transform(X_tfidf_abs)


results = []

for k in range(2, 11):

    model = KMeans(n_clusters=k, random_state=42, n_init=20)

    labels = model.fit_predict(X_abs_svd)

    results.append(
        {
            "K": k,
            "Silhouette": silhouette_score(X_abs_svd, labels),
            "Inertia": model.inertia_,
        }
    )


k_results = pd.DataFrame(results)

display(k_results)


# ============================================================
# 12. Silhouette para elegir número de clusters
# ============================================================

plt.figure(figsize=(8, 5))

sns.lineplot(data=k_results, x="K", y="Silhouette", marker="o")

plt.xlabel("Número de clusters")
plt.ylabel("Silhouette score")
plt.title("Selección del número de clusters temáticos")

plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Network de papers basado en similitud NLP de abstracts
#
# Nodo:
#   - un paper
#
# Arista:
#   - similitud coseno entre abstracts
#
# Peso:
#   - Similarity
#
# Usa X_tfidf_abs ya calculado anteriormente
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity

# ============================================================
# 1. Calcular similitud coseno entre todos los abstracts
# ============================================================

similarity_matrix = cosine_similarity(X_tfidf_abs)

print("Forma matriz:", similarity_matrix.shape)


# ============================================================
# 2. Ver distribución de similitudes
#    Solo triángulo superior, sin diagonal
# ============================================================

upper_triangle = similarity_matrix[np.triu_indices_from(similarity_matrix, k=1)]

print(pd.Series(upper_triangle).describe())


plt.figure(figsize=(9, 5))

plt.hist(upper_triangle, bins=50, alpha=0.8)

plt.xlabel("Similitud coseno")
plt.ylabel("Número de pares de papers")
plt.title("Distribución de similitud entre abstracts")

plt.tight_layout()
plt.show()

# ============================================================
# 3. Crear network
#
# Umbral inicial:
# conectar papers con similitud >= 0.30
# ============================================================

similarity_threshold = 0.25

G_papers = nx.Graph()


# ------------------------------------------------------------
# Nodos
# ------------------------------------------------------------

for i, row in df.reset_index(drop=True).iterrows():

    G_papers.add_node(
        i,
        PMID=str(row["PMID"]),
        Title=row["Title"],
        Year=row["Year"] if pd.notna(row["Year"]) else "",
        Abstract_Cluster=(
            int(row["Abstract_Cluster"])
            if "Abstract_Cluster" in df.columns and pd.notna(row["Abstract_Cluster"])
            else -1
        ),
    )


# ------------------------------------------------------------
# Aristas
# ------------------------------------------------------------

n = len(df)

for i in range(n):

    for j in range(i + 1, n):

        similarity = similarity_matrix[i, j]

        if similarity >= similarity_threshold:

            G_papers.add_edge(i, j, Weight=float(similarity))


print("Papers:", G_papers.number_of_nodes())
print("Relaciones:", G_papers.number_of_edges())


# ============================================================
# 4. Eliminar papers aislados para el plot
# ============================================================

connected_nodes = [node for node in G_papers.nodes() if G_papers.degree(node) > 0]

G_papers_plot = G_papers.subgraph(connected_nodes).copy()


print("Papers conectados:", G_papers_plot.number_of_nodes())

print("Relaciones:", G_papers_plot.number_of_edges())
```

```python
# ============================================================
# 5. Métricas básicas de la red
# ============================================================

degree = dict(G_papers_plot.degree())

weighted_degree = dict(G_papers_plot.degree(weight="Weight"))

nx.set_node_attributes(G_papers_plot, degree, "Degree")

nx.set_node_attributes(G_papers_plot, weighted_degree, "WeightedDegree")


paper_network_stats = (
    pd.DataFrame(
        [
            {
                "Node": node,
                "PMID": G_papers_plot.nodes[node]["PMID"],
                "Title": G_papers_plot.nodes[node]["Title"],
                "Year": G_papers_plot.nodes[node]["Year"],
                "Cluster": G_papers_plot.nodes[node]["Abstract_Cluster"],
                "Degree": degree[node],
                "WeightedDegree": weighted_degree[node],
            }
            for node in G_papers_plot.nodes()
        ]
    )
    .sort_values("WeightedDegree", ascending=False)
    .reset_index(drop=True)
)


display(paper_network_stats.head(30))
```

```python
# ============================================================
# 6. Plot global
#
# Tamaño = grado ponderado
# Color = cluster NLP previo
# Grosor de arista = similitud
# ============================================================

pos_papers = nx.spring_layout(
    G_papers_plot, weight="Weight", k=0.7, iterations=200, seed=42
)


node_sizes = [
    60 + G_papers_plot.nodes[node]["WeightedDegree"] * 80
    for node in G_papers_plot.nodes()
]


node_colors = [
    G_papers_plot.nodes[node]["Abstract_Cluster"] for node in G_papers_plot.nodes()
]


edge_widths = [
    0.2 + G_papers_plot[u][v]["Weight"] * 3 for u, v in G_papers_plot.edges()
]


plt.figure(figsize=(18, 18))


nx.draw_networkx_edges(
    G_papers_plot, pos_papers, width=edge_widths, edge_color="gray", alpha=0.15
)


nodes_plot = nx.draw_networkx_nodes(
    G_papers_plot,
    pos_papers,
    node_size=node_sizes,
    node_color=node_colors,
    cmap="tab10",
    alpha=0.85,
)


plt.colorbar(nodes_plot, label="Cluster NLP", shrink=0.6)

plt.title(
    f"Network de papers por similitud de abstracts\n"
    f"Similitud coseno ≥ {similarity_threshold}",
    fontsize=16,
)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 7. Detectar comunidades directamente en el network NLP
# ============================================================

communities_papers = nx.community.louvain_communities(
    G_papers_plot, weight="Weight", resolution=1.0, seed=42
)


paper_community_map = {}

for community_id, community in enumerate(communities_papers):

    for node in community:

        paper_community_map[node] = community_id


nx.set_node_attributes(G_papers_plot, paper_community_map, "NLP_Community")


print("Número de comunidades:", len(communities_papers))

# ============================================================
# 8. Plot coloreado por comunidades detectadas
# ============================================================

community_colors = [
    G_papers_plot.nodes[node]["NLP_Community"] for node in G_papers_plot.nodes()
]


plt.figure(figsize=(18, 18))


nx.draw_networkx_edges(
    G_papers_plot, pos_papers, width=edge_widths, edge_color="gray", alpha=0.12
)


nodes_plot = nx.draw_networkx_nodes(
    G_papers_plot,
    pos_papers,
    node_size=node_sizes,
    node_color=community_colors,
    cmap="tab20",
    alpha=0.9,
)


plt.colorbar(nodes_plot, label="Comunidad NLP", shrink=0.6)

plt.title("Comunidades de papers según similitud semántica de abstracts", fontsize=16)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 9. Etiquetar solamente papers centrales
# ============================================================

central_threshold = paper_network_stats["WeightedDegree"].quantile(0.90)


central_nodes = paper_network_stats.loc[
    paper_network_stats["WeightedDegree"] >= central_threshold, "Node"
]


labels = {node: G_papers_plot.nodes[node]["PMID"] for node in central_nodes}


plt.figure(figsize=(18, 18))


nx.draw_networkx_edges(
    G_papers_plot, pos_papers, width=edge_widths, edge_color="gray", alpha=0.12
)


nx.draw_networkx_nodes(
    G_papers_plot,
    pos_papers,
    node_size=node_sizes,
    node_color=community_colors,
    cmap="tab20",
    alpha=0.9,
)


nx.draw_networkx_labels(
    G_papers_plot, pos_papers, labels=labels, font_size=8, font_weight="bold"
)


plt.title("Network NLP de papers — papers centrales etiquetados", fontsize=16)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 10. Relaciones más fuertes entre papers
# ============================================================

paper_relations = (
    pd.DataFrame(
        [
            {
                "PMID1": G_papers_plot.nodes[u]["PMID"],
                "Title1": G_papers_plot.nodes[u]["Title"],
                "PMID2": G_papers_plot.nodes[v]["PMID"],
                "Title2": G_papers_plot.nodes[v]["Title"],
                "Similarity": data["Weight"],
            }
            for u, v, data in G_papers_plot.edges(data=True)
        ]
    )
    .sort_values("Similarity", ascending=False)
    .reset_index(drop=True)
)


display(paper_relations.head(50))

# ============================================================
# 11. Estadísticas de comunidades
# ============================================================

community_records = []

for node in G_papers_plot.nodes():

    community_records.append(
        {
            "Node": node,
            "PMID": G_papers_plot.nodes[node]["PMID"],
            "Title": G_papers_plot.nodes[node]["Title"],
            "Year": G_papers_plot.nodes[node]["Year"],
            "Community": G_papers_plot.nodes[node]["NLP_Community"],
            "Degree": G_papers_plot.nodes[node]["Degree"],
            "WeightedDegree": G_papers_plot.nodes[node]["WeightedDegree"],
        }
    )


paper_communities_df = pd.DataFrame(community_records)


community_stats = (
    paper_communities_df.groupby("Community")
    .agg(
        N_Papers=("PMID", "count"),
        Mean_Degree=("Degree", "mean"),
        Mean_WeightedDegree=("WeightedDegree", "mean"),
    )
    .sort_values("N_Papers", ascending=False)
    .reset_index()
)


display(community_stats)
```

```python
# ============================================================
# 12. Guardar network y resultados
# ============================================================

nx.write_gexf(G_papers_plot, os.path.join("ml_pubmed_ifibio", "paper_nlp_network.gexf"))

paper_network_stats.to_pickle(
    os.path.join("ml_pubmed_ifibio", "paper_nlp_network_stats.pkl")
)

paper_relations.to_pickle(os.path.join("ml_pubmed_ifibio", "paper_nlp_relations.pkl"))

paper_communities_df.to_pickle(
    os.path.join("ml_pubmed_ifibio", "paper_nlp_communities.pkl")
)

community_stats.to_pickle(
    os.path.join("ml_pubmed_ifibio", "paper_nlp_community_stats.pkl")
)


import pickle

with open(os.path.join("ml_pubmed_ifibio", "paper_nlp_network.pkl"), "wb") as f:
    pickle.dump(G_papers_plot, f)
```

```python
# ============================================================
# 1. DATASET INTEGRADO POR PAPER
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx
import ast

ml_df = df[["PMID", "Title", "Year"]].copy()

ml_df["PMID"] = ml_df["PMID"].astype(str)

print(ml_df.shape)
display(ml_df.head())
```

```python
# ============================================================
# 2. FEATURES NLP DEL ABSTRACT
#
# Usamos las componentes SVD calculadas anteriormente
# ============================================================

for i in range(X_abs_svd.shape[1]):

    ml_df[f"NLP_{i+1}"] = X_abs_svd[:, i]


print("Features NLP agregadas:", X_abs_svd.shape[1])
```

```python
# ============================================================
# 3. CLUSTER NLP DEL ABSTRACT
# ============================================================

if "Abstract_Cluster" in df.columns:

    ml_df["Abstract_Cluster"] = df["Abstract_Cluster"].values
```

```python
# ============================================================
# 4. FEATURES DEL NETWORK NLP DE PAPERS
#
# Degree
# WeightedDegree
# PageRank
# Betweenness
# Clustering coefficient
# ============================================================

paper_degree = dict(G_papers_plot.degree())

paper_weighted_degree = dict(G_papers_plot.degree(weight="Weight"))

paper_pagerank = nx.pagerank(G_papers_plot, weight="Weight")


# Para betweenness, similitud alta = distancia corta

G_temp = G_papers_plot.copy()

for u, v in G_temp.edges():

    G_temp[u][v]["Distance"] = 1 / G_temp[u][v]["Weight"]


paper_betweenness = nx.betweenness_centrality(
    G_temp, weight="Distance", normalized=True
)


paper_clustering = nx.clustering(G_papers_plot, weight="Weight")
```

```python
# ============================================================
# 5. Pasar métricas NLP network a PMID
# ============================================================

paper_network_features = []

for node in G_papers_plot.nodes():

    paper_network_features.append(
        {
            "PMID": str(G_papers_plot.nodes[node]["PMID"]),
            "Paper_Degree": paper_degree[node],
            "Paper_WeightedDegree": paper_weighted_degree[node],
            "Paper_PageRank": paper_pagerank[node],
            "Paper_Betweenness": paper_betweenness[node],
            "Paper_Clustering": paper_clustering[node],
            "NLP_Community": G_papers_plot.nodes[node].get("NLP_Community", -1),
        }
    )


paper_network_features = pd.DataFrame(paper_network_features)


ml_df = ml_df.merge(paper_network_features, on="PMID", how="left")
```

```python
# ============================================================
# 6. FEATURES DE AUTORES POR PAPER
# ============================================================

authors_parsed = df["Authors"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


paper_author_features = []


for pmid, author_list in zip(df["PMID"], authors_parsed):

    names = [
        author.get("AuthorClean") for author in author_list if author.get("AuthorClean")
    ]

    names = list(dict.fromkeys(names))

    # Autores presentes en network de coautoría

    valid_authors = [author for author in names if author in G]

    n_authors = len(names)

    n_ifibio = sum(G.nodes[a].get("IFIBIO", False) for a in valid_authors)

    publications = [G.nodes[a].get("N_Publications", 0) for a in valid_authors]

    degrees = [G.degree(a) for a in valid_authors]

    weighted_degrees = [G.degree(a, weight="Weight") for a in valid_authors]

    paper_author_features.append(
        {
            "PMID": str(pmid),
            "N_Authors": n_authors,
            "N_IFIBIO_Authors": n_ifibio,
            "Fraction_IFIBIO": n_ifibio / n_authors if n_authors > 0 else 0,
            "Mean_Author_Publications": np.mean(publications) if publications else 0,
            "Max_Author_Publications": np.max(publications) if publications else 0,
            "Mean_Author_Degree": np.mean(degrees) if degrees else 0,
            "Max_Author_Degree": np.max(degrees) if degrees else 0,
            "Mean_Author_WeightedDegree": (
                np.mean(weighted_degrees) if weighted_degrees else 0
            ),
            "Max_Author_WeightedDegree": (
                np.max(weighted_degrees) if weighted_degrees else 0
            ),
        }
    )


paper_author_features = pd.DataFrame(paper_author_features)


ml_df = ml_df.merge(paper_author_features, on="PMID", how="left")
```

```python
# ============================================================
# 7. DIVERSIDAD DE AFILIACIONES POR PAPER
# ============================================================

affiliation_features = []


for pmid, author_list in zip(df["PMID"], authors_parsed):

    affiliations = []

    for author in author_list:

        for affiliation in author.get("Affiliations", []):

            text = affiliation.get("Affiliation")

            if text:
                affiliations.append(text.strip())

    affiliations = list(dict.fromkeys(affiliations))

    affiliation_features.append(
        {"PMID": str(pmid), "N_Affiliations": len(affiliations)}
    )


affiliation_features = pd.DataFrame(affiliation_features)


ml_df = ml_df.merge(affiliation_features, on="PMID", how="left")
```

```python
# ============================================================
# 8. FEATURES MeSH + KEYWORDS
# ============================================================

mesh_parsed = df["MeSH"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)

keywords_parsed = df["Keywords"].apply(
    lambda x: (
        ast.literal_eval(x)
        if isinstance(x, str)
        else (x if isinstance(x, list) else [])
    )
)


ml_df["N_MeSH"] = mesh_parsed.apply(len)

ml_df["N_Keywords"] = keywords_parsed.apply(len)
```

```python
# ============================================================
# 9. LONGITUD DEL ABSTRACT
# ============================================================

ml_df["Abstract_Words"] = df["Tokens"].apply(len).values
```

```python
# ============================================================
# 10. FEATURES TEMPORALES
# ============================================================

ml_df["Year"] = pd.to_numeric(ml_df["Year"], errors="coerce")


ml_df["Years_From_First"] = ml_df["Year"] - ml_df["Year"].min()
```

```python
# ============================================================
# 11. AGREGAR CLUSTER SOCIAL DE LOS AUTORES
#
# Cluster dominante de coautoría para cada paper
# ============================================================

paper_social_clusters = []


for pmid, author_list in zip(df["PMID"], authors_parsed):

    names = [
        author.get("AuthorClean") for author in author_list if author.get("AuthorClean")
    ]

    social_clusters = [
        G.nodes[a].get("Cluster")
        for a in names
        if (a in G and G.nodes[a].get("Cluster") is not None)
    ]

    if social_clusters:

        dominant_cluster = pd.Series(social_clusters).value_counts().index[0]

        n_social_clusters = len(set(social_clusters))

    else:

        dominant_cluster = -1
        n_social_clusters = 0

    paper_social_clusters.append(
        {
            "PMID": str(pmid),
            "Social_Cluster": dominant_cluster,
            "N_Social_Clusters": n_social_clusters,
        }
    )


paper_social_clusters = pd.DataFrame(paper_social_clusters)


ml_df = ml_df.merge(paper_social_clusters, on="PMID", how="left")
```

```python
# ============================================================
# 12. INSPECCIONAR DATASET FINAL
# ============================================================

print("Papers:", ml_df.shape[0])

print("Features:", ml_df.shape[1])

display(ml_df.head())

display(ml_df.describe(include="all").T)
```

```python
# ============================================================
# 14. MATRIZ PARA MACHINE LEARNING
# ============================================================

from sklearn.preprocessing import StandardScaler

exclude_columns = [
    "PMID",
    "Title",
    "Abstract_Cluster",
    "NLP_Community",
    "Social_Cluster",
]


feature_columns = [
    column
    for column in ml_df.columns
    if (column not in exclude_columns and pd.api.types.is_numeric_dtype(ml_df[column]))
]


X_ml = ml_df[feature_columns].replace([np.inf, -np.inf], np.nan).fillna(0)


print("Número de features:", len(feature_columns))

print(feature_columns)
```

```python
# ============================================================
# 15. ESTANDARIZAR
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_ml)

# ============================================================
# 16. PCA MULTIDIMENSIONAL
# ============================================================

from sklearn.decomposition import PCA

pca = PCA()

X_pca = pca.fit_transform(X_scaled)


explained = pca.explained_variance_ratio_ * 100


cumulative = np.cumsum(explained)


plt.figure(figsize=(9, 5))

plt.plot(range(1, len(cumulative) + 1), cumulative, marker="o")

plt.axhline(80, linestyle="--")

plt.xlabel("Número de componentes")

plt.ylabel("Varianza acumulada (%)")

plt.title("Varianza explicada — dataset integrado")

plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 17. PCA 2D
# ============================================================

pca2 = PCA(n_components=2)

X_pca2 = pca2.fit_transform(X_scaled)


pca_df = pd.DataFrame(
    {
        "PC1": X_pca2[:, 0],
        "PC2": X_pca2[:, 1],
        "Year": ml_df["Year"],
        "PMID": ml_df["PMID"],
    }
)


plt.figure(figsize=(11, 8))

scatter = plt.scatter(
    pca_df["PC1"], pca_df["PC2"], c=pca_df["Year"], cmap="viridis", s=55, alpha=0.75
)


plt.colorbar(scatter, label="Año")

plt.xlabel(f"PC1 ({pca2.explained_variance_ratio_[0]*100:.1f}%)")

plt.ylabel(f"PC2 ({pca2.explained_variance_ratio_[1]*100:.1f}%)")

plt.title("PCA — información integrada de papers")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 1. DATASET AUTOR × PAPER
#
# Vincular cada autor con las features integradas de sus papers
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

author_paper_records = []

for paper in paper_data:

    pmid = str(paper["PMID"])

    for author in paper["Authors"]:

        author_paper_records.append({"PMID": pmid, "Author": author})


author_paper_df = pd.DataFrame(author_paper_records)


author_paper_ml = author_paper_df.merge(ml_df, on="PMID", how="left")


print(author_paper_ml.shape)

display(author_paper_ml.head())

# ============================================================
# 2. FEATURES QUE VAN A DEFINIR EL PERFIL DE CADA AUTOR
# ============================================================

nlp_features = [col for col in ml_df.columns if col.startswith("NLP_")]


paper_features = [
    "Paper_Degree",
    "Paper_WeightedDegree",
    "Paper_PageRank",
    "Paper_Betweenness",
    "Paper_Clustering",
    "N_Authors",
    "N_IFIBIO_Authors",
    "Fraction_IFIBIO",
    "N_Affiliations",
    "N_MeSH",
    "N_Keywords",
    "Abstract_Words",
]


profile_features = nlp_features + [
    col for col in paper_features if col in ml_df.columns
]


print("Features usadas:", len(profile_features))
print(profile_features)
```

```python
# ============================================================
# 3. PERFIL PROMEDIO DE CADA AUTOR
# ============================================================

author_profiles = author_paper_ml.groupby("Author")[profile_features].mean()


# Número de papers
author_npapers = (
    author_paper_ml.groupby("Author")["PMID"].nunique().rename("N_Publications")
)


# Primer y último año
author_year_stats = author_paper_ml.groupby("Author")["Year"].agg(
    First_Year="min", Last_Year="max", Mean_Year="mean"
)


author_summary = pd.concat([author_npapers, author_year_stats], axis=1)


display(author_summary.sort_values("N_Publications", ascending=False).head(30))

# ============================================================
# 4. ATRIBUTOS DE LA RED ORIGINAL DE AUTORES
# ============================================================

author_network_attributes = []


for author in author_summary.index:

    if author in G:

        author_network_attributes.append(
            {
                "Author": author,
                "IFIBIO": G.nodes[author].get("IFIBIO", False),
                "Original_Degree": G.degree(author),
                "Original_WeightedDegree": G.degree(author, weight="Weight"),
                "Social_Cluster": G.nodes[author].get("Cluster", -1),
            }
        )

    else:

        author_network_attributes.append(
            {
                "Author": author,
                "IFIBIO": False,
                "Original_Degree": 0,
                "Original_WeightedDegree": 0,
                "Social_Cluster": -1,
            }
        )


author_network_attributes = pd.DataFrame(author_network_attributes).set_index("Author")


author_summary = author_summary.join(author_network_attributes)


display(author_summary.head())

# ============================================================
# 5. CLUSTER MULTIDIMENSIONAL DOMINANTE DE CADA AUTOR
# ============================================================

author_integrated_cluster = (
    author_paper_ml.dropna(subset=["Integrated_Cluster"])
    .groupby("Author")["Integrated_Cluster"]
    .agg(lambda x: x.value_counts().index[0])
    .rename("Integrated_Cluster")
)


author_summary = author_summary.join(author_integrated_cluster)


author_summary["Integrated_Cluster"] = (
    author_summary["Integrated_Cluster"].fillna(-1).astype(int)
)


# ============================================================
# 6. SIMILITUD ENTRE LOS PERFILES DE AUTORES
# ============================================================

X_author = author_profiles.replace([np.inf, -np.inf], np.nan).fillna(0)


scaler_author = StandardScaler()

X_author_scaled = scaler_author.fit_transform(X_author)


author_similarity = cosine_similarity(X_author_scaled)


author_names = X_author.index.to_numpy()


print(author_similarity.shape)

# ============================================================
# 7. FUNCIÓN DE SOLAPAMIENTO TEMPORAL
#
# 1 = carreras científicas completamente solapadas
# 0 = sin solapamiento temporal
# ============================================================

temporal_overlap = {}


for i, author1 in enumerate(author_names):

    y1_start = author_summary.loc[author1, "First_Year"]

    y1_end = author_summary.loc[author1, "Last_Year"]

    for j in range(i + 1, len(author_names)):

        author2 = author_names[j]

        y2_start = author_summary.loc[author2, "First_Year"]

        y2_end = author_summary.loc[author2, "Last_Year"]

        if pd.isna(y1_start) or pd.isna(y1_end) or pd.isna(y2_start) or pd.isna(y2_end):

            overlap = 0

        else:

            overlap_years = max(0, min(y1_end, y2_end) - max(y1_start, y2_start) + 1)

            total_years = max(y1_end, y2_end) - min(y1_start, y2_start) + 1

            overlap = overlap_years / total_years

        temporal_overlap[(author1, author2)] = overlap
```

```python
# ============================================================
# 8. CREAR NETWORK MULTIDIMENSIONAL DE AUTORES
# ============================================================

G_multi = nx.Graph()


# ------------------------------------------------------------
# NODOS
# ------------------------------------------------------------

for author in author_names:

    row = author_summary.loc[author]

    G_multi.add_node(
        author,
        N_Publications=int(row["N_Publications"]),
        IFIBIO=bool(row["IFIBIO"]),
        First_Year=float(row["First_Year"]) if pd.notna(row["First_Year"]) else -1,
        Last_Year=float(row["Last_Year"]) if pd.notna(row["Last_Year"]) else -1,
        Mean_Year=float(row["Mean_Year"]) if pd.notna(row["Mean_Year"]) else -1,
        Social_Cluster=int(row["Social_Cluster"]),
        Integrated_Cluster=int(row["Integrated_Cluster"]),
    )
```

```python
# ============================================================
# 9. CREAR ARISTAS MULTIDIMENSIONALES
# ============================================================

semantic_threshold = 0.40


for i in range(len(author_names)):

    author1 = author_names[i]

    for j in range(i + 1, len(author_names)):

        author2 = author_names[j]

        # ------------------------------------
        # Coautoría real
        # ------------------------------------

        if author1 in G and author2 in G and G.has_edge(author1, author2):

            coauthorship = G[author1][author2].get("Weight", 0)

        else:

            coauthorship = 0

        # ------------------------------------
        # Similitud temática/multidimensional
        # ------------------------------------

        semantic = float(author_similarity[i, j])

        # ------------------------------------
        # Solapamiento temporal
        # ------------------------------------

        temporal = temporal_overlap.get((author1, author2), 0)

        # ------------------------------------
        # Agregar si:
        #
        # existe colaboración REAL
        # o similitud >= threshold
        # ------------------------------------

        if coauthorship > 0 or semantic >= semantic_threshold:

            G_multi.add_edge(
                author1,
                author2,
                Coauthorship=float(coauthorship),
                SemanticSimilarity=semantic,
                TemporalOverlap=float(temporal),
            )
```

```python
# ============================================================
# 10. NORMALIZAR COAUTORÍA
# ============================================================

coauthor_values = [data["Coauthorship"] for _, _, data in G_multi.edges(data=True)]


max_coauthor = max(coauthor_values) if coauthor_values else 1


for u, v, data in G_multi.edges(data=True):

    data["CoauthorshipNorm"] = data["Coauthorship"] / max_coauthor
```

```python
# ============================================================
# 11. PESO MULTIDIMENSIONAL
#
# 50% colaboración real
# 35% similitud temática
# 15% coincidencia temporal
# ============================================================

for u, v, data in G_multi.edges(data=True):

    data["CompositeWeight"] = (
        0.50 * data["CoauthorshipNorm"]
        + 0.35 * data["SemanticSimilarity"]
        + 0.15 * data["TemporalOverlap"]
    )
```

```python
# ============================================================
# 12. ESTADÍSTICAS
# ============================================================

print("Autores:", G_multi.number_of_nodes())

print("Relaciones:", G_multi.number_of_edges())

print("Densidad:", nx.density(G_multi))
```

```python
# ============================================================
# 13. CENTRALIDADES MULTIDIMENSIONALES
# ============================================================

multi_degree = dict(G_multi.degree())


multi_weighted_degree = dict(G_multi.degree(weight="CompositeWeight"))


multi_pagerank = nx.pagerank(G_multi, weight="CompositeWeight")


G_distance = G_multi.copy()


for u, v in G_distance.edges():

    weight = G_distance[u][v]["CompositeWeight"]

    G_distance[u][v]["Distance"] = 1 / max(weight, 1e-8)


multi_betweenness = nx.betweenness_centrality(
    G_distance, weight="Distance", normalized=True
)

# ============================================================
# 14. TABLA DE CENTRALIDADES
# ============================================================

multi_stats = pd.DataFrame(
    {
        "Author": list(G_multi.nodes()),
        "N_Publications": [G_multi.nodes[a]["N_Publications"] for a in G_multi.nodes()],
        "IFIBIO": [G_multi.nodes[a]["IFIBIO"] for a in G_multi.nodes()],
        "Degree": [multi_degree[a] for a in G_multi.nodes()],
        "WeightedDegree": [multi_weighted_degree[a] for a in G_multi.nodes()],
        "PageRank": [multi_pagerank[a] for a in G_multi.nodes()],
        "Betweenness": [multi_betweenness[a] for a in G_multi.nodes()],
    }
)


multi_stats = multi_stats.sort_values("WeightedDegree", ascending=False).reset_index(
    drop=True
)


display(multi_stats.head(30))

# ============================================================
# 15. CLUSTERS MULTIDIMENSIONALES DE AUTORES
# ============================================================

multi_communities = nx.community.louvain_communities(
    G_multi, weight="CompositeWeight", resolution=1.0, seed=42
)


multi_cluster_map = {}


for cluster_id, community in enumerate(multi_communities):

    for author in community:

        multi_cluster_map[author] = cluster_id


nx.set_node_attributes(G_multi, multi_cluster_map, "MultiCluster")


print("Clusters multidimensionales:", len(multi_communities))


print(
    "Modularidad:",
    nx.community.modularity(G_multi, multi_communities, weight="CompositeWeight"),
)

# ============================================================
# 16. PLOT DEL NETWORK MULTIDIMENSIONAL
# ============================================================

pos_multi = nx.spring_layout(
    G_multi, weight="CompositeWeight", k=0.7, iterations=250, seed=42
)


clusters = [G_multi.nodes[a]["MultiCluster"] for a in G_multi.nodes()]


node_sizes = [50 + G_multi.nodes[a]["N_Publications"] * 25 for a in G_multi.nodes()]


edge_widths = [0.15 + G_multi[u][v]["CompositeWeight"] * 3 for u, v in G_multi.edges()]


plt.figure(figsize=(20, 20))


nx.draw_networkx_edges(
    G_multi, pos_multi, width=edge_widths, edge_color="gray", alpha=0.10
)


nodes_plot = nx.draw_networkx_nodes(
    G_multi,
    pos_multi,
    node_size=node_sizes,
    node_color=clusters,
    cmap="tab20",
    alpha=0.85,
)


plt.colorbar(nodes_plot, label="Cluster multidimensional", shrink=0.6)


plt.title(
    "Network multidimensional de autores\n"
    "Coautoría + similitud científica + temporalidad"
)


plt.axis("off")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# 17. PLOT DESTACANDO IFIBIO
# ============================================================

ifibio_multi = [a for a in G_multi.nodes() if G_multi.nodes[a]["IFIBIO"]]


external_multi = [a for a in G_multi.nodes() if not G_multi.nodes[a]["IFIBIO"]]


plt.figure(figsize=(20, 20))


nx.draw_networkx_edges(
    G_multi, pos_multi, width=edge_widths, edge_color="gray", alpha=0.07
)


# Externos
nx.draw_networkx_nodes(
    G_multi,
    pos_multi,
    nodelist=external_multi,
    node_size=[30 + G_multi.nodes[a]["N_Publications"] * 12 for a in external_multi],
    node_color="gray",
    alpha=0.12,
)


# IFIBIO
ifibio_plot = nx.draw_networkx_nodes(
    G_multi,
    pos_multi,
    nodelist=ifibio_multi,
    node_size=[100 + G_multi.nodes[a]["N_Publications"] * 35 for a in ifibio_multi],
    node_color=[G_multi.nodes[a]["MultiCluster"] for a in ifibio_multi],
    cmap="tab20",
    alpha=0.95,
)


# --------------------------------------------
# Etiquetar IFIBIO por encima de la mediana
# --------------------------------------------

median_ifibio_papers = np.median(
    [G_multi.nodes[a]["N_Publications"] for a in ifibio_multi]
)


labels = {
    a: a
    for a in ifibio_multi
    if G_multi.nodes[a]["N_Publications"] > median_ifibio_papers
}


nx.draw_networkx_labels(G_multi, pos_multi, labels=labels, font_size=8)


plt.title("Network multidimensional de autores — IFIBIO")


plt.axis("off")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# 17. PLOT DESTACANDO IFIBIO
# - conexiones IFIBIO-IFIBIO más visibles
# - labels más grandes para los 10 autores IFIBIO más prolíficos
# ============================================================

ifibio_multi = [a for a in G_multi.nodes() if G_multi.nodes[a]["IFIBIO"]]

external_multi = [a for a in G_multi.nodes() if not G_multi.nodes[a]["IFIBIO"]]


# ------------------------------------------------------------
# Separar conexiones
# ------------------------------------------------------------

ifibio_edges = []
other_edges = []

for u, v in G_multi.edges():

    if G_multi.nodes[u]["IFIBIO"] and G_multi.nodes[v]["IFIBIO"]:
        ifibio_edges.append((u, v))

    else:
        other_edges.append((u, v))


# Grosor de conexiones generales
other_widths = [0.1 + G_multi[u][v]["CompositeWeight"] * 1.5 for u, v in other_edges]


# Grosor aumentado para conexiones IFIBIO-IFIBIO
ifibio_widths = [0.8 + G_multi[u][v]["CompositeWeight"] * 5 for u, v in ifibio_edges]


# ------------------------------------------------------------
# N_top autores IFIBIO más prolíficos
# ------------------------------------------------------------
N_top = 20

topN_ifibio = sorted(
    ifibio_multi, key=lambda a: G_multi.nodes[a]["N_Publications"], reverse=True
)[:N_top]


# Resto de autores IFIBIO a etiquetar
other_ifibio_labels = [
    a
    for a in ifibio_multi
    if (
        G_multi.nodes[a]["N_Publications"]
        > np.median([G_multi.nodes[x]["N_Publications"] for x in ifibio_multi])
        and a not in topN_ifibio
    )
]


# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(20, 20))


# ------------------------------------------------------------
# Conexiones generales
# ------------------------------------------------------------

nx.draw_networkx_edges(
    G_multi,
    pos_multi,
    edgelist=other_edges,
    width=other_widths,
    edge_color="gray",
    alpha=0.05,
)


# ------------------------------------------------------------
# Conexiones IFIBIO-IFIBIO
# ------------------------------------------------------------

nx.draw_networkx_edges(
    G_multi,
    pos_multi,
    edgelist=ifibio_edges,
    width=ifibio_widths,
    edge_color="tab:gray",
    alpha=0.40,
)


# ------------------------------------------------------------
# Autores externos
# ------------------------------------------------------------

nx.draw_networkx_nodes(
    G_multi,
    pos_multi,
    nodelist=external_multi,
    node_size=[30 + G_multi.nodes[a]["N_Publications"] * 12 for a in external_multi],
    node_color="gray",
    alpha=0.10,
)


# ------------------------------------------------------------
# Autores IFIBIO
# ------------------------------------------------------------

ifibio_plot = nx.draw_networkx_nodes(
    G_multi,
    pos_multi,
    nodelist=ifibio_multi,
    node_size=[100 + G_multi.nodes[a]["N_Publications"] * 35 for a in ifibio_multi],
    node_color=[G_multi.nodes[a]["MultiCluster"] for a in ifibio_multi],
    cmap="tab20",
    alpha=0.95,
    linewidths=0.1,
    edgecolors="black",
)


# ------------------------------------------------------------
# Labels IFIBIO secundarios
# ------------------------------------------------------------

labels_regular = {a: a for a in other_ifibio_labels}

nx.draw_networkx_labels(
    G_multi, pos_multi, labels=labels_regular, font_size=8, alpha=0.8
)


# ------------------------------------------------------------
# Labels TOP 10
# ------------------------------------------------------------

labels_topN = {a: a for a in topN_ifibio}

nx.draw_networkx_labels(
    G_multi, pos_multi, labels=labels_topN, font_size=14, font_weight="bold"
)


plt.title("Network multidimensional de autores — IFIBIO", fontsize=18)

plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 17. PLOT DESTACANDO IFIBIO
# - mayor separación entre nodos
# - conexiones IFIBIO-IFIBIO más visibles
# - labels más legibles
# - labels grandes para los N autores IFIBIO más prolíficos
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

# ------------------------------------------------------------
# Autores IFIBIO / externos
# ------------------------------------------------------------

ifibio_multi = [a for a in G_multi.nodes() if G_multi.nodes[a]["IFIBIO"]]

external_multi = [a for a in G_multi.nodes() if not G_multi.nodes[a]["IFIBIO"]]


# ------------------------------------------------------------
# Recalcular layout con mayor separación
# ------------------------------------------------------------

pos_multi_plot = nx.spring_layout(
    G_multi,
    weight="CompositeWeight",
    k=1.8,  # aumentar para separar nodos
    iterations=500,
    seed=42,
    scale=3,
)


# ------------------------------------------------------------
# Separar conexiones
# ------------------------------------------------------------

ifibio_edges = []
other_edges = []

for u, v in G_multi.edges():

    if G_multi.nodes[u]["IFIBIO"] and G_multi.nodes[v]["IFIBIO"]:
        ifibio_edges.append((u, v))

    else:
        other_edges.append((u, v))


# ------------------------------------------------------------
# Grosor de conexiones
# ------------------------------------------------------------

other_widths = [0.05 + G_multi[u][v]["CompositeWeight"] * 0.8 for u, v in other_edges]

ifibio_widths = [0.8 + G_multi[u][v]["CompositeWeight"] * 4 for u, v in ifibio_edges]


# ------------------------------------------------------------
# Top N autores IFIBIO más prolíficos
# ------------------------------------------------------------

N_top = 20

topN_ifibio = sorted(
    ifibio_multi, key=lambda a: G_multi.nodes[a]["N_Publications"], reverse=True
)[:N_top]


# ------------------------------------------------------------
# Mediana publicaciones IFIBIO
# ------------------------------------------------------------

median_ifibio_papers = np.median(
    [G_multi.nodes[a]["N_Publications"] for a in ifibio_multi]
)


# ------------------------------------------------------------
# Resto de autores IFIBIO a etiquetar
# ------------------------------------------------------------

other_ifibio_labels = [
    a
    for a in ifibio_multi
    if (
        G_multi.nodes[a]["N_Publications"] > median_ifibio_papers
        and a not in topN_ifibio
    )
]


# ------------------------------------------------------------
# Posiciones separadas para labels
# ------------------------------------------------------------

label_pos_regular = {
    a: (pos_multi_plot[a][0], pos_multi_plot[a][1] + 0.05) for a in other_ifibio_labels
}


label_pos_top = {
    a: (pos_multi_plot[a][0], pos_multi_plot[a][1] + 0.08) for a in topN_ifibio
}


# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(24, 24))


# ------------------------------------------------------------
# Conexiones generales
# ------------------------------------------------------------

nx.draw_networkx_edges(
    G_multi,
    pos_multi_plot,
    edgelist=other_edges,
    width=other_widths,
    edge_color="gray",
    alpha=0.025,
)


# ------------------------------------------------------------
# Conexiones IFIBIO-IFIBIO
# ------------------------------------------------------------

nx.draw_networkx_edges(
    G_multi,
    pos_multi_plot,
    edgelist=ifibio_edges,
    width=ifibio_widths,
    edge_color="dimgray",
    alpha=0.55,
)


# ------------------------------------------------------------
# Autores externos
# ------------------------------------------------------------

nx.draw_networkx_nodes(
    G_multi,
    pos_multi_plot,
    nodelist=external_multi,
    node_size=[15 + G_multi.nodes[a]["N_Publications"] * 5 for a in external_multi],
    node_color="lightgray",
    alpha=0.10,
    linewidths=0,
)


# ------------------------------------------------------------
# Autores IFIBIO
# ------------------------------------------------------------

ifibio_plot = nx.draw_networkx_nodes(
    G_multi,
    pos_multi_plot,
    nodelist=ifibio_multi,
    node_size=[120 + G_multi.nodes[a]["N_Publications"] * 45 for a in ifibio_multi],
    node_color=[G_multi.nodes[a]["MultiCluster"] for a in ifibio_multi],
    cmap="tab20",
    alpha=0.95,
    linewidths=0.8,
    edgecolors="black",
)


# ------------------------------------------------------------
# Labels secundarios
# ------------------------------------------------------------

labels_regular = {a: a for a in other_ifibio_labels}

nx.draw_networkx_labels(
    G_multi,
    label_pos_regular,
    labels=labels_regular,
    font_size=9,
    font_weight="normal",
    alpha=0.85,
    bbox=dict(facecolor="white", edgecolor="none", alpha=0.55, pad=0.3),
)


# ------------------------------------------------------------
# Labels TOP N
# ------------------------------------------------------------

labels_topN = {a: a for a in topN_ifibio}

nx.draw_networkx_labels(
    G_multi,
    label_pos_top,
    labels=labels_topN,
    font_size=15,
    font_weight="bold",
    bbox=dict(facecolor="white", edgecolor="none", alpha=0.75, pad=0.4),
)


# ------------------------------------------------------------
# Título
# ------------------------------------------------------------

plt.title("Network multidimensional de autores — IFIBIO", fontsize=22, pad=20)

plt.axis("off")
plt.tight_layout()

plt.show()
```

```python
# ============================================================
# 2. DEFINIR AÑO DE CORTE
#
# Features: todo lo publicado hasta cutoff_year
# Target: publicaciones en los 3 años siguientes
# ============================================================

cutoff_year = 2023
future_window = 3

future_start = cutoff_year + 1
future_end = cutoff_year + future_window

print("Datos históricos hasta:", cutoff_year)
print("Endpoint:", future_start, "-", future_end)
```

```python
# ============================================================
# 3. AUTOR × PAPER × AÑO
# ============================================================

author_year_records = []

for pmid, year, author_list in zip(df["PMID"], df["Year"], authors_parsed):

    if pd.isna(year):
        continue

    for author in author_list:

        name = author.get("AuthorClean")

        if name:

            author_year_records.append(
                {"Author": name, "PMID": str(pmid), "Year": int(year)}
            )


author_year_df = pd.DataFrame(author_year_records)

display(author_year_df.head())


# ============================================================
# 4. FEATURES HISTÓRICAS BÁSICAS
# ============================================================

historical = author_year_df[author_year_df["Year"] <= cutoff_year].copy()


historical_features = historical.groupby("Author").agg(
    Papers_Historical=("PMID", "nunique"),
    First_Year=("Year", "min"),
    Last_Year=("Year", "max"),
)


historical_features["Career_Length"] = (
    cutoff_year - historical_features["First_Year"] + 1
)


display(historical_features.sort_values("Papers_Historical", ascending=False).head(30))
```

```python
# ============================================================
# 5. PRODUCTIVIDAD EN LOS ÚLTIMOS 1, 3 Y 5 AÑOS
# ============================================================

papers_last1 = (
    historical[historical["Year"] >= cutoff_year].groupby("Author")["PMID"].nunique()
)


papers_last3 = (
    historical[historical["Year"] >= cutoff_year - 2]
    .groupby("Author")["PMID"]
    .nunique()
)


papers_last5 = (
    historical[historical["Year"] >= cutoff_year - 4]
    .groupby("Author")["PMID"]
    .nunique()
)


historical_features["Papers_Last1"] = papers_last1

historical_features["Papers_Last3"] = papers_last3

historical_features["Papers_Last5"] = papers_last5


historical_features[["Papers_Last1", "Papers_Last3", "Papers_Last5"]] = (
    historical_features[["Papers_Last1", "Papers_Last3", "Papers_Last5"]].fillna(0)
)

# ============================================================
# 6. ENDPOINT:
# PUBLICACIONES EN LOS 3 AÑOS SIGUIENTES
# ============================================================

future = author_year_df[
    (author_year_df["Year"] >= future_start) & (author_year_df["Year"] <= future_end)
]


future_productivity = future.groupby("Author")["PMID"].nunique().rename("Papers_Next3")


model_df = historical_features.join(future_productivity)


model_df["Papers_Next3"] = model_df["Papers_Next3"].fillna(0).astype(int)


model_df = model_df.reset_index()


display(model_df.sort_values("Papers_Next3", ascending=False).head(30))

# ============================================================
# 7. DISTRIBUCIÓN DEL ENDPOINT
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(data=model_df, x="Papers_Next3", discrete=True)

plt.xlabel("Publicaciones en los siguientes 3 años")

plt.ylabel("Autores")

plt.title("Distribución del endpoint")

plt.tight_layout()
plt.show()


print(model_df["Papers_Next3"].describe())

print("\nAutores sin publicaciones futuras:", (model_df["Papers_Next3"] == 0).sum())

print("Autores con publicaciones futuras:", (model_df["Papers_Next3"] > 0).sum())
```

```python
# ============================================================
# 1. CONFIGURACIÓN TEMPORAL
# ============================================================

cutoff_year = 2022
future_window = 3

future_start = cutoff_year + 1
future_end = cutoff_year + future_window

print("Features hasta:", cutoff_year)
print("Endpoint:", future_start, "-", future_end)
```

```python
# ============================================================
# 2. PAPERS HISTÓRICOS
# ============================================================

historical_papers = df[df["Year"] <= cutoff_year].copy()

historical_pmids = set(historical_papers["PMID"].astype(str))

print("Papers históricos:", len(historical_papers))

# ============================================================
# 3. RED DE COAUTORÍA HISTÓRICA
# ============================================================

from itertools import combinations
import networkx as nx
import numpy as np
import pandas as pd

G_hist = nx.Graph()


for pmid, year, author_list in zip(df["PMID"], df["Year"], authors_parsed):

    if pd.isna(year) or year > cutoff_year:
        continue

    names = [a.get("AuthorClean") for a in author_list if a.get("AuthorClean")]

    names = list(dict.fromkeys(names))

    # nodos
    for author in names:

        if author not in G_hist:

            G_hist.add_node(author, N_Publications=0)

        G_hist.nodes[author]["N_Publications"] += 1

    # aristas
    for a1, a2 in combinations(sorted(names), 2):

        if G_hist.has_edge(a1, a2):

            G_hist[a1][a2]["Weight"] += 1

        else:

            G_hist.add_edge(a1, a2, Weight=1)


print("Autores:", G_hist.number_of_nodes())
print("Colaboraciones:", G_hist.number_of_edges())
```

```python
# ============================================================
# 4. CENTRALIDADES HISTÓRICAS
# ============================================================

degree_hist = dict(G_hist.degree())

weighted_degree_hist = dict(G_hist.degree(weight="Weight"))

pagerank_hist = nx.pagerank(G_hist, weight="Weight")


clustering_hist = nx.clustering(G_hist, weight="Weight")


G_hist_distance = G_hist.copy()

for u, v in G_hist_distance.edges():

    G_hist_distance[u][v]["Distance"] = 1 / G_hist_distance[u][v]["Weight"]


betweenness_hist = nx.betweenness_centrality(
    G_hist_distance, weight="Distance", normalized=True
)

# ============================================================
# 5. FEATURES LOCALES DEL NETWORK
# ============================================================

triangles_hist = nx.triangles(G_hist)


network_records = []


for author in G_hist.nodes():

    neighbors = set(G_hist.neighbors(author))

    second_neighbors = set()

    for neighbor in neighbors:

        second_neighbors.update(G_hist.neighbors(neighbor))

    second_neighbors.discard(author)

    second_neighbors -= neighbors

    network_records.append(
        {
            "Author": author,
            "Net_Degree": degree_hist[author],
            "Net_WeightedDegree": weighted_degree_hist[author],
            "Net_PageRank": pagerank_hist[author],
            "Net_Betweenness": betweenness_hist[author],
            "Net_Clustering": clustering_hist[author],
            "Net_Triangles": triangles_hist[author],
            "Net_SecondNeighbors": len(second_neighbors),
        }
    )


network_features = pd.DataFrame(network_records)


display(network_features.sort_values("Net_PageRank", ascending=False).head(20))
```

```python
# ============================================================
# 6. NLP HISTÓRICO POR AUTOR
# ============================================================

nlp_columns = [f"NLP_{i+1}" for i in range(X_abs_svd.shape[1])]


nlp_papers = pd.DataFrame(X_abs_svd, columns=nlp_columns)


nlp_papers["PMID"] = df["PMID"].astype(str).values

nlp_papers["Year"] = df["Year"].values


nlp_papers = nlp_papers[nlp_papers["Year"] <= cutoff_year]

# ============================================================
# 7. VINCULAR NLP CON AUTORES
# ============================================================

author_paper_hist = author_year_df[author_year_df["Year"] <= cutoff_year][
    ["Author", "PMID", "Year"]
].copy()


author_paper_hist["PMID"] = author_paper_hist["PMID"].astype(str)


author_nlp = author_paper_hist.merge(nlp_papers, on=["PMID", "Year"], how="left")


# ============================================================
# 8. PERFIL NLP PROMEDIO DEL AUTOR
# ============================================================

author_nlp_features = author_nlp.groupby("Author")[nlp_columns].mean().reset_index()


display(author_nlp_features.head())

# ============================================================
# 9. PRODUCTIVIDAD HISTÓRICA
# ============================================================

productivity_features = historical_features.reset_index()


display(productivity_features.head())
```

```python
# ============================================================
# 10. DATASET MULTIDIMENSIONAL
# ============================================================

model_multi_df = productivity_features.merge(
    network_features, on="Author", how="left"
).merge(author_nlp_features, on="Author", how="left")
```

```python
# ============================================================
# 11. ENDPOINT
# ============================================================

future_productivity = (
    author_year_df[
        (author_year_df["Year"] >= future_start)
        & (author_year_df["Year"] <= future_end)
    ]
    .groupby("Author")["PMID"]
    .nunique()
    .rename("Papers_Next3")
)


model_multi_df = (
    model_multi_df.set_index("Author").join(future_productivity).reset_index()
)


model_multi_df["Papers_Next3"] = model_multi_df["Papers_Next3"].fillna(0)


model_multi_df["Active_Next3"] = (model_multi_df["Papers_Next3"] > 0).astype(int)


print(model_multi_df["Active_Next3"].value_counts())

display(model_multi_df.head())
```

```python
# ============================================================
# 12. FEATURES
# ============================================================

productivity_columns = [
    "Papers_Historical",
    "Career_Length",
    "Papers_Last1",
    "Papers_Last3",
    "Papers_Last5",
]


network_columns = [
    "Net_Degree",
    "Net_WeightedDegree",
    "Net_PageRank",
    "Net_Betweenness",
    "Net_Clustering",
    "Net_Triangles",
    "Net_SecondNeighbors",
]


features_multi = productivity_columns + network_columns + nlp_columns


X = model_multi_df[features_multi].replace([np.inf, -np.inf], np.nan).fillna(0)


y = model_multi_df["Active_Next3"]


print("Autores:", len(X))
print("Features:", len(features_multi))
```

```python
# ============================================================
# 13. TRAIN / TEST
# ============================================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)


print("Train:", len(X_train))
print("Test :", len(X_test))


# ============================================================
# 14. BALANCE DE CLASES
# ============================================================

negative = (y_train == 0).sum()

positive = (y_train == 1).sum()


scale_pos_weight = negative / positive


print("Negativos:", negative)
print("Positivos:", positive)

print("scale_pos_weight:", round(scale_pos_weight, 3))
```

```python
# ============================================================
# 15. XGBOOST MULTIDIMENSIONAL
# ============================================================

from xgboost import XGBClassifier

xgb_multi = XGBClassifier(
    n_estimators=500,
    max_depth=3,
    learning_rate=0.03,
    min_child_weight=3,
    subsample=0.8,
    colsample_bytree=0.7,
    reg_alpha=0.2,
    reg_lambda=2.0,
    scale_pos_weight=scale_pos_weight,
    eval_metric="logloss",
    random_state=42,
)


xgb_multi.fit(X_train, y_train)

# ============================================================
# 16. PREDICCIÓN
# ============================================================

pred_multi = xgb_multi.predict(X_test)


prob_multi = xgb_multi.predict_proba(X_test)[:, 1]
```

```python
# ============================================================
# PROBABILIDAD DE PUBLICAR EN LOS PRÓXIMOS N AÑOS
# 20 AUTORES MÁS PROLÍFICOS
#
# Endpoint para cada N:
#   1 = publica >= 1 paper entre cutoff+1 y cutoff+N
#   0 = no publica
#
# Usa las features multidimensionales ya construidas:
# productividad + network + NLP
# ============================================================

from xgboost import XGBClassifier
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# CONFIGURACIÓN
# ------------------------------------------------------------

cutoff_year = 2020

horizons = [1, 2, 3, 4, 5]

N_top = 20


# ------------------------------------------------------------
# 20 AUTORES MÁS PROLÍFICOS HASTA EL CUTOFF
# ------------------------------------------------------------

top20_authors = (
    model_multi_df.sort_values("Papers_Historical", ascending=False)
    .head(N_top)["Author"]
    .tolist()
)


# ------------------------------------------------------------
# MATRIZ DE FEATURES
# ------------------------------------------------------------

X_all = model_multi_df[features_multi].replace([np.inf, -np.inf], np.nan).fillna(0)


# ------------------------------------------------------------
# TABLA DE RESULTADOS
# ------------------------------------------------------------

probability_table = model_multi_df[["Author", "Papers_Historical"]].copy()


# ============================================================
# ENTRENAR UN MODELO PARA CADA HORIZONTE
# ============================================================

for N in horizons:

    future_end = cutoff_year + N

    # --------------------------------------------------------
    # Endpoint real para este horizonte
    # --------------------------------------------------------

    future_counts = (
        author_year_df[
            (author_year_df["Year"] > cutoff_year)
            & (author_year_df["Year"] <= future_end)
        ]
        .groupby("Author")["PMID"]
        .nunique()
    )

    y_N = model_multi_df["Author"].map(future_counts).fillna(0).gt(0).astype(int)

    # --------------------------------------------------------
    # Balance de clases
    # --------------------------------------------------------

    negative = (y_N == 0).sum()
    positive = (y_N == 1).sum()

    scale_pos_weight_N = negative / positive if positive > 0 else 1

    # --------------------------------------------------------
    # XGBOOST
    # --------------------------------------------------------

    model_N = XGBClassifier(
        n_estimators=500,
        max_depth=3,
        learning_rate=0.03,
        min_child_weight=3,
        subsample=0.8,
        colsample_bytree=0.7,
        reg_alpha=0.2,
        reg_lambda=2.0,
        scale_pos_weight=scale_pos_weight_N,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
    )

    model_N.fit(X_all, y_N)

    # --------------------------------------------------------
    # PROBABILIDAD
    # --------------------------------------------------------

    probability_table[f"Prob_{N}y"] = model_N.predict_proba(X_all)[:, 1]


# ============================================================
# QUEDARSE CON TOP 20
# ============================================================

top20_probability = (
    probability_table[probability_table["Author"].isin(top20_authors)]
    .sort_values("Papers_Historical", ascending=False)
    .reset_index(drop=True)
)


display(top20_probability.round(3))


# ============================================================
# TABLA EN PORCENTAJES
# ============================================================

top20_percent = top20_probability.copy()


prob_columns = [f"Prob_{N}y" for N in horizons]


top20_percent[prob_columns] = (top20_percent[prob_columns] * 100).round(1)


top20_percent.columns = [
    "Autor",
    "Publicaciones históricas",
    "1 año",
    "2 años",
    "3 años",
    "4 años",
    "5 años",
]


display(top20_percent)

# ============================================================
# HEATMAP
# ============================================================

import seaborn as sns
import matplotlib.pyplot as plt

heatmap_data = top20_percent.set_index("Autor")[
    ["1 año", "2 años", "3 años", "4 años", "5 años"]
]


plt.figure(figsize=(10, 11))


sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".1f",
    cmap="viridis",
    vmin=0,
    vmax=100,
    linewidths=0.5,
    cbar_kws={"label": "Probabilidad de publicar (%)"},
)


plt.xlabel("Horizonte de predicción")

plt.ylabel("")

plt.title("Probabilidad de actividad científica futura\n" "20 autores más prolíficos")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# CURVAS POR AUTOR
# ============================================================

plot_df = top20_probability.melt(
    id_vars=["Author", "Papers_Historical"],
    value_vars=prob_columns,
    var_name="Horizon",
    value_name="Probability",
)


plot_df["Years"] = plot_df["Horizon"].str.extract(r"(\d+)").astype(int)


plt.figure(figsize=(12, 9))


for author in top20_authors:

    data_author = plot_df[plot_df["Author"] == author].sort_values("Years")

    plt.plot(
        data_author["Years"],
        data_author["Probability"],
        marker="o",
        linewidth=2,
        alpha=0.75,
        label=author,
    )


plt.xlabel("Próximos N años")

plt.ylabel("Probabilidad de publicar")

plt.ylim(0, 1.05)

plt.xticks(horizons)

plt.title("Probabilidad de actividad científica futura\n" "20 autores más prolíficos")

plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=8)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# 4. JACCARD
# ============================================================

jaccard = {
    tuple(sorted((u, v))): score
    for u, v, score in nx.jaccard_coefficient(G, candidate_pairs)
}


# ============================================================
# 5. ADAMIC-ADAR
# ============================================================

adamic_adar = {
    tuple(sorted((u, v))): score
    for u, v, score in nx.adamic_adar_index(G, candidate_pairs)
}

# ============================================================
# 6. RESOURCE ALLOCATION
# ============================================================

resource_allocation = {
    tuple(sorted((u, v))): score
    for u, v, score in nx.resource_allocation_index(G, candidate_pairs)
}

# ============================================================
# 7. VECINOS COMUNES
# ============================================================

common_neighbors = {}

for u, v in candidate_pairs:

    common_neighbors[tuple(sorted((u, v)))] = len(list(nx.common_neighbors(G, u, v)))

# ============================================================
# 8. DATASET DE COLABORACIONES POTENCIALES
# ============================================================

recommendations = []


for author_ifibio, candidate in candidate_pairs:

    pair = tuple(sorted((author_ifibio, candidate)))

    semantic_similarity = similarity_lookup.get(pair, 0)

    temporal = temporal_overlap.get(pair, 0)

    recommendations.append(
        {
            "IFIBIO_Author": author_ifibio,
            "Candidate": candidate,
            "Candidate_IFIBIO": G_multi.nodes[candidate].get("IFIBIO", False),
            "SemanticSimilarity": semantic_similarity,
            "TemporalOverlap": temporal,
            "CommonNeighbors": common_neighbors.get(pair, 0),
            "Jaccard": jaccard.get(pair, 0),
            "AdamicAdar": adamic_adar.get(pair, 0),
            "ResourceAllocation": resource_allocation.get(pair, 0),
            "Candidate_Publications": G_multi.nodes[candidate].get("N_Publications", 0),
        }
    )


recommendations_df = pd.DataFrame(recommendations)

display(recommendations_df.head())

# ============================================================
# 9. NORMALIZACIÓN
# ============================================================

from sklearn.preprocessing import MinMaxScaler

score_columns = [
    "SemanticSimilarity",
    "TemporalOverlap",
    "CommonNeighbors",
    "Jaccard",
    "AdamicAdar",
    "ResourceAllocation",
]


scaler_affinity = MinMaxScaler()


recommendations_df[[col + "_Norm" for col in score_columns]] = (
    scaler_affinity.fit_transform(recommendations_df[score_columns])
)
```

```python
# ============================================================
# 10. SCORE DE AFINIDAD
# ============================================================

recommendations_df["AffinityScore"] = (
    0.50 * recommendations_df["SemanticSimilarity_Norm"]
    + 0.15 * recommendations_df["TemporalOverlap_Norm"]
    + 0.10 * recommendations_df["Jaccard_Norm"]
    + 0.10 * recommendations_df["AdamicAdar_Norm"]
    + 0.10 * recommendations_df["ResourceAllocation_Norm"]
    + 0.05 * recommendations_df["CommonNeighbors_Norm"]
)


recommendations_df["AffinityScore_Percent"] = recommendations_df["AffinityScore"] * 100

# ============================================================
# 11. TOP 5 COLABORADORES PARA CADA AUTOR IFIBIO
# ============================================================

top5_collaborators = (
    recommendations_df.sort_values(
        ["IFIBIO_Author", "AffinityScore"], ascending=[True, False]
    )
    .groupby("IFIBIO_Author")
    .head(5)
    .reset_index(drop=True)
)


display(
    top5_collaborators[
        [
            "IFIBIO_Author",
            "Candidate",
            "Candidate_IFIBIO",
            "AffinityScore_Percent",
            "SemanticSimilarity",
            "TemporalOverlap",
            "CommonNeighbors",
            "Jaccard",
            "AdamicAdar",
        ]
    ].round(3)
)
```

```python
# ============================================================
# RECONSTRUIR paper_mesh CORRECTAMENTE
# DESDE df["MeSH"]
# ============================================================

import ast
import pandas as pd

mesh_parsed = df["MeSH"].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) and x.strip() else []
)


paper_mesh = []


for mesh_list in mesh_parsed:

    terms = []

    for item in mesh_list:

        if isinstance(item, dict):

            term = (
                item.get("DescriptorName")
                or item.get("Descriptor")
                or item.get("MeSH")
                or item.get("Term")
                or item.get("Name")
            )

        else:

            term = item

        if term:

            terms.append(str(term).strip())

    # eliminar duplicados dentro del paper
    terms = list(dict.fromkeys(terms))

    paper_mesh.append(terms)


print("Papers:", len(paper_mesh))

print("Ejemplos:")

for x in paper_mesh[:10]:
    print(x)
```

```python
# ============================================================
# 1. RED DE COOCURRENCIA MeSH
# ============================================================

import networkx as nx
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from itertools import combinations

G_thematic = nx.Graph()


for terms in paper_mesh:

    # eliminar duplicados dentro del mismo paper
    terms = list(dict.fromkeys(terms))

    # nodos
    for term in terms:

        if term not in G_thematic:

            G_thematic.add_node(term, Frequency=0)

        G_thematic.nodes[term]["Frequency"] += 1

    # coocurrencias
    for term1, term2 in combinations(sorted(terms), 2):

        if G_thematic.has_edge(term1, term2):

            G_thematic[term1][term2]["Weight"] += 1

        else:

            G_thematic.add_edge(term1, term2, Weight=1)


print("Términos MeSH:", G_thematic.number_of_nodes())

print("Coocurrencias:", G_thematic.number_of_edges())
```

```python
# ============================================================
# 2. FILTRAR TÉRMINOS RAROS
# ============================================================

min_frequency = 3


selected_terms = [
    term
    for term in G_thematic.nodes()
    if G_thematic.nodes[term]["Frequency"] >= min_frequency
]


G_thematic_filtered = G_thematic.subgraph(selected_terms).copy()


print("MeSH retenidos:", G_thematic_filtered.number_of_nodes())

print("Coocurrencias retenidas:", G_thematic_filtered.number_of_edges())
```

```python
mesh_frequency_df = pd.DataFrame(
    {
        "MeSH": list(G_thematic_filtered.nodes()),
        "Frequency": [
            G_thematic_filtered.nodes[x]["Frequency"]
            for x in G_thematic_filtered.nodes()
        ],
    }
).sort_values("Frequency", ascending=False)


display(mesh_frequency_df.head(30))
```

```python
# ============================================================
# 3. COMUNIDADES TEMÁTICAS
# ============================================================

communities = nx.community.louvain_communities(
    G_thematic_filtered, weight="Weight", resolution=1.0, seed=42
)


print("Número de temas:", len(communities))
```

```python
theme_map = {}


for theme_id, community in enumerate(communities):

    for term in community:

        theme_map[term] = theme_id


nx.set_node_attributes(G_thematic_filtered, theme_map, "Theme")
```

```python
# ============================================================
# 4. DENSIDAD Y CENTRALIDAD
# ============================================================

theme_records = []


for theme_id, community in enumerate(communities):

    nodes_theme = set(community)

    internal_weight = 0

    external_weight = 0

    for node in nodes_theme:

        for neighbor in G_thematic_filtered.neighbors(node):

            weight = G_thematic_filtered[node][neighbor]["Weight"]

            if neighbor in nodes_theme:

                internal_weight += weight

            else:

                external_weight += weight

    # cada edge interno fue contado dos veces
    internal_weight /= 2

    n_nodes = len(nodes_theme)

    # densidad normalizada por tamaño
    if n_nodes > 1:

        density = internal_weight / (n_nodes * (n_nodes - 1) / 2)

    else:

        density = 0

    # centralidad externa
    centrality = external_weight / n_nodes

    # frecuencia total del tema
    frequency = sum(
        G_thematic_filtered.nodes[node]["Frequency"] for node in nodes_theme
    )

    theme_records.append(
        {
            "Theme": theme_id,
            "N_Terms": n_nodes,
            "Frequency": frequency,
            "Density": density,
            "Centrality": centrality,
        }
    )


theme_df = pd.DataFrame(theme_records)


display(theme_df)
```

```python
# ============================================================
# 5. TÉRMINOS REPRESENTATIVOS
# ============================================================

theme_labels = {}


for theme_id, community in enumerate(communities):

    terms_ranked = sorted(
        community, key=lambda x: G_thematic_filtered.nodes[x]["Frequency"], reverse=True
    )

    top_terms = terms_ranked[:3]

    theme_labels[theme_id] = " / ".join(top_terms)


theme_df["Label"] = theme_df["Theme"].map(theme_labels)


display(
    theme_df[
        ["Theme", "Label", "N_Terms", "Frequency", "Density", "Centrality"]
    ].sort_values("Frequency", ascending=False)
)
```

```python
# ============================================================
# 6. CENTROS DEL THEMATIC MAP
# ============================================================

centrality_center = theme_df["Centrality"].median()


density_center = theme_df["Density"].median()


print("Centralidad media:", centrality_center)

print("Densidad media:", density_center)
```

```python
# ============================================================
# 7. CLASIFICACIÓN DE LOS TEMAS
# ============================================================

conditions = [
    (
        (theme_df["Centrality"] >= centrality_center)
        & (theme_df["Density"] >= density_center)
    ),
    (
        (theme_df["Centrality"] >= centrality_center)
        & (theme_df["Density"] < density_center)
    ),
    (
        (theme_df["Centrality"] < centrality_center)
        & (theme_df["Density"] >= density_center)
    ),
    (
        (theme_df["Centrality"] < centrality_center)
        & (theme_df["Density"] < density_center)
    ),
]


categories = ["Motor", "Basic", "Niche", "Emerging / Declining"]


theme_df["Type"] = np.select(conditions, categories, default="Other")


display(
    theme_df[["Label", "Type", "Frequency", "Centrality", "Density"]].sort_values(
        "Frequency", ascending=False
    )
)
```

```python
# ============================================================
# 15. CRECIMIENTO RECIENTE
# ============================================================

last_year = int(df["Year"].max())

recent_start = last_year - 3

previous_start = last_year - 9
previous_end = last_year - 5


print("Período anterior:", previous_start, "-", previous_end)

print("Período reciente:", recent_start, "-", last_year)
```

```python
# ============================================================
# 16. COMPARAR LOS DOS PERÍODOS
# ============================================================

theme_growth_records = []


for theme, community in enumerate(communities):

    label = theme_labels[theme]

    previous = paper_theme_df[
        (paper_theme_df["Theme"] == theme)
        & (paper_theme_df["Year"] >= previous_start)
        & (paper_theme_df["Year"] <= previous_end)
    ]["PMID"].nunique()

    recent = paper_theme_df[
        (paper_theme_df["Theme"] == theme)
        & (paper_theme_df["Year"] >= recent_start)
        & (paper_theme_df["Year"] <= last_year)
    ]["PMID"].nunique()

    growth = (recent - previous) / (previous + 1)

    theme_growth_records.append(
        {
            "Theme": theme,
            "Label": label,
            "Previous_Papers": previous,
            "Recent_Papers": recent,
            "Growth": growth,
        }
    )


theme_growth = pd.DataFrame(theme_growth_records)

# ============================================================
# 17. EMERGENCE SCORE
# ============================================================

theme_growth["EmergenceScore"] = theme_growth["Growth"] * np.log1p(
    theme_growth["Recent_Papers"]
)


theme_growth = theme_growth.sort_values("EmergenceScore", ascending=False).reset_index(
    drop=True
)


display(theme_growth.round(3))
```

```python
# ============================================================
# 18. CLASIFICAR TRAYECTORIA
# ============================================================

theme_growth["Trajectory"] = "Stable"


theme_growth.loc[
    (theme_growth["Growth"] > 0.30) & (theme_growth["Recent_Papers"] >= 3), "Trajectory"
] = "Emerging"


theme_growth.loc[(theme_growth["Growth"] < -0.30), "Trajectory"] = "Declining"


display(
    theme_growth[
        [
            "Label",
            "Previous_Papers",
            "Recent_Papers",
            "Growth",
            "EmergenceScore",
            "Trajectory",
        ]
    ].round(3)
)
```

```python
# ============================================================
# 39. DEPENDENCIA DE COLABORADORES RECURRENTES
# ============================================================

collaboration_dependency = []

for author in ifibio_authors:

    if author not in G:
        continue

    neighbors = list(G.neighbors(author))

    weights = np.array([G[author][n]["Weight"] for n in neighbors])

    if len(weights) == 0:
        continue

    total = weights.sum()

    proportions = weights / total

    # Herfindahl-Hirschman Index
    hhi = np.sum(proportions**2)

    strongest = weights.max() / total

    collaboration_dependency.append(
        {
            "Author": author,
            "N_Collaborators": len(neighbors),
            "Total_Coauthorships": total,
            "Collaboration_HHI": hhi,
            "Strongest_Collaborator_Share": strongest,
        }
    )


collaboration_dependency = pd.DataFrame(collaboration_dependency)

collaboration_dependency = collaboration_dependency.merge(
    network_stats[["Author", "N_Publications"]], on="Author", how="left"
)

collaboration_dependency = collaboration_dependency[
    collaboration_dependency["N_Publications"] >= 3
].sort_values("Collaboration_HHI", ascending=False)

display(
    collaboration_dependency[
        [
            "Author",
            "N_Publications",
            "N_Collaborators",
            "Collaboration_HHI",
            "Strongest_Collaborator_Share",
        ]
    ]
    .head(20)
    .round(3)
)

plt.figure(figsize=(11, 8))

sns.scatterplot(
    data=collaboration_dependency,
    x="N_Collaborators",
    y="Collaboration_HHI",
    size="N_Publications",
    sizes=(80, 1000),
    alpha=0.75,
    edgecolor="black",
)

for _, row in collaboration_dependency.iterrows():

    plt.text(
        row["N_Collaborators"],
        row["Collaboration_HHI"],
        row["Author"],
        fontsize=7,
        ha="left",
        va="bottom",
    )

plt.xlabel("Número de colaboradores únicos")
plt.ylabel("Concentración de colaboraciones (HHI)")
plt.title("Diversificación de la red de colaboración — IFIBIO")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 40. COLABORACIÓN DENTRO Y FUERA DEL CLUSTER
# ============================================================

cluster_collaboration = []

for author in ifibio_authors:

    if author not in G:
        continue

    author_cluster = cluster_map.get(author)

    internal_weight = 0
    external_weight = 0

    internal_people = 0
    external_people = 0

    for neighbor in G.neighbors(author):

        weight = G[author][neighbor]["Weight"]

        if cluster_map.get(neighbor) == author_cluster:

            internal_weight += weight
            internal_people += 1

        else:

            external_weight += weight
            external_people += 1

    total_weight = internal_weight + external_weight

    external_fraction = external_weight / total_weight if total_weight > 0 else 0

    cluster_collaboration.append(
        {
            "Author": author,
            "Internal_Coauthorships": internal_weight,
            "External_Coauthorships": external_weight,
            "Internal_Collaborators": internal_people,
            "External_Collaborators": external_people,
            "CrossClusterFraction": external_fraction,
        }
    )


cluster_collaboration = pd.DataFrame(cluster_collaboration)

cluster_collaboration = cluster_collaboration.merge(
    network_stats[["Author", "N_Publications"]], on="Author", how="left"
)

cluster_collaboration = cluster_collaboration[
    cluster_collaboration["N_Publications"] >= 3
].sort_values("CrossClusterFraction", ascending=False)

display(
    cluster_collaboration[
        [
            "Author",
            "N_Publications",
            "Internal_Collaborators",
            "External_Collaborators",
            "CrossClusterFraction",
        ]
    ]
    .head(20)
    .round(3)
)

plot_data = cluster_collaboration.head(20).sort_values(
    "CrossClusterFraction", ascending=True
)

plt.figure(figsize=(10, 8))

sns.barplot(data=plot_data, x="CrossClusterFraction", y="Author")

plt.xlabel("Fracción de colaboración fuera del cluster")
plt.ylabel("")

plt.title("Investigadores IFIBIO que más conectan comunidades")

plt.xlim(0, 1)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# TEMPORAL GRAPHSAGE — LINK PREDICTION DE NUEVAS COLABORACIONES
#
# Datos disponibles: 2014–2026
#
# Entrenamiento:
#   snapshots 2017–2021
#
# Validación:
#   red hasta 2022 -> colaboraciones nuevas 2023–2025
#
# Test temporal final:
#   red hasta 2023 -> colaboraciones nuevas 2024–2026
#
# El modelo intenta responder:
# P(nueva colaboración en próximos 3 años | historia hasta cutoff)
#
# IMPORTANTE:
# una "colaboración nueva" = dos autores que nunca habían publicado
# juntos hasta el cutoff y que publican juntos durante el horizonte futuro.
# ============================================================

# Si hace falta:
# !pip install -q torch-geometric

import ast
import random
import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

import torch
import torch.nn.functional as F

from torch import nn
from torch_geometric.nn import SAGEConv

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    confusion_matrix,
)

from sklearn.preprocessing import StandardScaler

# ============================================================
# CONFIGURACIÓN
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)


START_YEAR = 2014
END_YEAR = 2026

HORIZON = 3

TRAIN_CUTOFFS = [2017, 2018, 2019, 2020, 2021]

VAL_CUTOFF = 2022
TEST_CUTOFF = 2023

NEGATIVE_RATIO = 3

MIN_PAPERS_AUTHOR = 1
```

```python
# ============================================================
# 1. PARSEAR AUTORES
# ============================================================

if "authors_parsed" not in globals():

    authors_parsed = df["Authors"].apply(
        lambda x: (
            ast.literal_eval(x)
            if isinstance(x, str) and x.strip()
            else (x if isinstance(x, list) else [])
        )
    )


# usamos AuthorClean ya corregido anteriormente

paper_records = []

for pmid, year, author_list in zip(df["PMID"], df["Year"], authors_parsed):

    if pd.isna(year):
        continue

    year = int(year)

    if year < START_YEAR or year > END_YEAR:
        continue

    names = [
        author.get("AuthorClean") for author in author_list if author.get("AuthorClean")
    ]

    names = list(dict.fromkeys(names))

    if len(names) == 0:
        continue

    paper_records.append({"PMID": str(pmid), "Year": year, "Authors": names})


papers_temporal = pd.DataFrame(paper_records)

print("Papers:", len(papers_temporal))

print("Período:", papers_temporal["Year"].min(), "-", papers_temporal["Year"].max())


# ============================================================
# 2. UNIVERSO DE AUTORES
# ============================================================

all_authors_gnn = sorted(
    {author for authors in papers_temporal["Authors"] for author in authors}
)

author_to_idx = {author: i for i, author in enumerate(all_authors_gnn)}

idx_to_author = {i: author for author, i in author_to_idx.items()}

N_AUTHORS = len(all_authors_gnn)

print("Autores:", N_AUTHORS)


# ============================================================
# 3. FUNCIÓN PARA CREAR SNAPSHOT TEMPORAL
# ============================================================


def crear_snapshot(cutoff, horizon=3, negative_ratio=3, training=True):

    # --------------------------------------------------------
    # HISTORIA DISPONIBLE HASTA CUTOFF
    # --------------------------------------------------------

    historical = papers_temporal[papers_temporal["Year"] <= cutoff].copy()

    future = papers_temporal[
        (papers_temporal["Year"] > cutoff)
        & (papers_temporal["Year"] <= cutoff + horizon)
    ].copy()

    # --------------------------------------------------------
    # GRAFO HISTÓRICO
    # --------------------------------------------------------

    G_hist = nx.Graph()

    G_hist.add_nodes_from(all_authors_gnn)

    for _, row in historical.iterrows():

        authors = row["Authors"]

        for i in range(len(authors)):

            for j in range(i + 1, len(authors)):

                a = authors[i]
                b = authors[j]

                if G_hist.has_edge(a, b):

                    G_hist[a][b]["weight"] += 1

                else:

                    G_hist.add_edge(a, b, weight=1)

    # --------------------------------------------------------
    # FEATURES HISTÓRICAS POR AUTOR
    # --------------------------------------------------------

    paper_counts = {author: 0 for author in all_authors_gnn}

    papers_last1 = {author: 0 for author in all_authors_gnn}

    papers_last3 = {author: 0 for author in all_authors_gnn}

    first_year = {author: None for author in all_authors_gnn}

    last_year = {author: None for author in all_authors_gnn}

    for _, row in historical.iterrows():

        year = row["Year"]

        for author in row["Authors"]:

            paper_counts[author] += 1

            if year >= cutoff:
                papers_last1[author] += 1

            if year >= cutoff - 2:
                papers_last3[author] += 1

            if first_year[author] is None:
                first_year[author] = year

            first_year[author] = min(first_year[author], year)

            if last_year[author] is None:
                last_year[author] = year

            last_year[author] = max(last_year[author], year)

    degree = dict(G_hist.degree())

    weighted_degree = dict(G_hist.degree(weight="weight"))

    pagerank = nx.pagerank(G_hist, weight="weight")

    clustering = nx.clustering(G_hist, weight="weight")

    features = []

    for author in all_authors_gnn:

        career_length = 0

        if first_year[author] is not None:

            career_length = cutoff - first_year[author] + 1

        features.append(
            [
                np.log1p(paper_counts[author]),
                np.log1p(papers_last1[author]),
                np.log1p(papers_last3[author]),
                np.log1p(career_length),
                np.log1p(degree.get(author, 0)),
                np.log1p(weighted_degree.get(author, 0)),
                pagerank.get(author, 0),
                clustering.get(author, 0),
            ]
        )

    x = np.array(features, dtype=np.float32)

    scaler = StandardScaler()

    x = scaler.fit_transform(x)

    x = torch.tensor(x, dtype=torch.float32)

    # --------------------------------------------------------
    # EDGE INDEX DEL GRAFO HISTÓRICO
    # --------------------------------------------------------

    edge_list = []

    for a, b in G_hist.edges():

        i = author_to_idx[a]
        j = author_to_idx[b]

        edge_list.append([i, j])

        edge_list.append([j, i])

    if len(edge_list) > 0:

        edge_index = torch.tensor(edge_list, dtype=torch.long).T.contiguous()

    else:

        edge_index = torch.empty((2, 0), dtype=torch.long)

    # --------------------------------------------------------
    # EDGES QUE APARECEN EN EL FUTURO
    # --------------------------------------------------------

    future_edges = set()

    for _, row in future.iterrows():

        authors = row["Authors"]

        for i in range(len(authors)):

            for j in range(i + 1, len(authors)):

                pair = tuple(sorted([authors[i], authors[j]]))

                future_edges.add(pair)

    # --------------------------------------------------------
    # POSITIVOS:
    # colaboraciones futuras que NO existían antes
    # --------------------------------------------------------

    positive_pairs = []

    for a, b in future_edges:

        if not G_hist.has_edge(a, b):

            # ambos deben existir científicamente
            # antes del cutoff

            if (
                paper_counts[a] >= MIN_PAPERS_AUTHOR
                and paper_counts[b] >= MIN_PAPERS_AUTHOR
            ):

                positive_pairs.append((author_to_idx[a], author_to_idx[b]))

    # --------------------------------------------------------
    # CANDIDATOS NEGATIVOS
    # --------------------------------------------------------

    active_authors = [
        author
        for author in all_authors_gnn
        if paper_counts[author] >= MIN_PAPERS_AUTHOR
    ]

    positive_set = {tuple(sorted(pair)) for pair in positive_pairs}

    negative_candidates = []

    for i in range(len(active_authors)):

        for j in range(i + 1, len(active_authors)):

            a = active_authors[i]
            b = active_authors[j]

            ia = author_to_idx[a]
            ib = author_to_idx[b]

            pair_idx = tuple(sorted((ia, ib)))

            # excluir colaboraciones históricas
            if G_hist.has_edge(a, b):
                continue

            # excluir positivos futuros
            if pair_idx in positive_set:
                continue

            negative_candidates.append(pair_idx)

    # --------------------------------------------------------
    # MUESTREO NEGATIVO
    # --------------------------------------------------------

    if training:

        n_negative = min(len(negative_candidates), len(positive_pairs) * negative_ratio)

        random.shuffle(negative_candidates)

        negative_pairs = negative_candidates[:n_negative]

    else:

        # para validación/test también mantenemos
        # una relación 1:3 para evaluación controlada

        n_negative = min(len(negative_candidates), len(positive_pairs) * negative_ratio)

        rng = random.Random(SEED + cutoff)

        negative_pairs = (
            rng.sample(negative_candidates, n_negative) if n_negative > 0 else []
        )

    # --------------------------------------------------------
    # EDGE LABEL INDEX + LABEL
    # --------------------------------------------------------

    pairs = positive_pairs + negative_pairs

    labels = [1] * len(positive_pairs) + [0] * len(negative_pairs)

    permutation = np.random.permutation(len(pairs))

    pairs = [pairs[i] for i in permutation]

    labels = [labels[i] for i in permutation]

    edge_label_index = torch.tensor(pairs, dtype=torch.long).T.contiguous()

    edge_label = torch.tensor(labels, dtype=torch.float32)

    return {
        "cutoff": cutoff,
        "x": x,
        "edge_index": edge_index,
        "edge_label_index": edge_label_index,
        "edge_label": edge_label,
        "G": G_hist,
        "positive_pairs": positive_pairs,
        "negative_pairs": negative_pairs,
        "paper_counts": paper_counts,
    }


# ============================================================
# 4. CREAR SNAPSHOTS
# ============================================================

train_snapshots = [
    crear_snapshot(
        cutoff, horizon=HORIZON, negative_ratio=NEGATIVE_RATIO, training=True
    )
    for cutoff in TRAIN_CUTOFFS
]


val_snapshot = crear_snapshot(
    VAL_CUTOFF, horizon=HORIZON, negative_ratio=NEGATIVE_RATIO, training=False
)


test_snapshot = crear_snapshot(
    TEST_CUTOFF, horizon=HORIZON, negative_ratio=NEGATIVE_RATIO, training=False
)


print("\nSNAPSHOTS\n")


for snapshot in train_snapshots + [val_snapshot, test_snapshot]:

    print(
        snapshot["cutoff"],
        "| positivos:",
        int(snapshot["edge_label"].sum()),
        "| negativos:",
        int((snapshot["edge_label"] == 0).sum()),
    )


# ============================================================
# 5. MODELO GRAPHSAGE
# ============================================================


class GraphSAGELinkPredictor(nn.Module):

    def __init__(self, in_channels, hidden_channels=64, embedding_channels=32):

        super().__init__()

        self.conv1 = SAGEConv(in_channels, hidden_channels)

        self.conv2 = SAGEConv(hidden_channels, embedding_channels)

        self.decoder = nn.Sequential(
            nn.Linear(embedding_channels * 4, 64),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(64, 1),
        )

    def encode(self, x, edge_index):

        x = self.conv1(x, edge_index)

        x = F.relu(x)

        x = F.dropout(x, p=0.2, training=self.training)

        x = self.conv2(x, edge_index)

        return x

    def decode(self, z, edge_label_index):

        src = z[edge_label_index[0]]

        dst = z[edge_label_index[1]]

        pair_features = torch.cat([src, dst, torch.abs(src - dst), src * dst], dim=1)

        logits = self.decoder(pair_features).squeeze()

        return logits

    def forward(self, x, edge_index, edge_label_index):

        z = self.encode(x, edge_index)

        logits = self.decode(z, edge_label_index)

        return logits


# ============================================================
# 6. INICIALIZAR
# ============================================================

n_features = train_snapshots[0]["x"].shape[1]


model_gnn = GraphSAGELinkPredictor(
    in_channels=n_features, hidden_channels=64, embedding_channels=32
).to(device)


optimizer = torch.optim.AdamW(model_gnn.parameters(), lr=0.002, weight_decay=1e-4)


criterion = nn.BCEWithLogitsLoss()


print(model_gnn)


# ============================================================
# 7. EVALUACIÓN
# ============================================================


def evaluar_snapshot(model, snapshot):

    model.eval()

    with torch.no_grad():

        x = snapshot["x"].to(device)

        edge_index = snapshot["edge_index"].to(device)

        edge_label_index = snapshot["edge_label_index"].to(device)

        y_true = snapshot["edge_label"].cpu().numpy()

        logits = model(x, edge_index, edge_label_index)

        probabilities = torch.sigmoid(logits).cpu().numpy()

    predictions = (probabilities >= 0.5).astype(int)

    metrics = {
        "ROC_AUC": roc_auc_score(y_true, probabilities),
        "PR_AUC": average_precision_score(y_true, probabilities),
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(y_true, predictions, zero_division=0),
        "Recall": recall_score(y_true, predictions, zero_division=0),
        "F1": f1_score(y_true, predictions, zero_division=0),
    }

    return (metrics, y_true, probabilities)


# ============================================================
# 8. ENTRENAMIENTO WALK-FORWARD
# ============================================================

N_EPOCHS = 300

history = []


best_val_auc = -np.inf

best_state = None


for epoch in range(1, N_EPOCHS + 1):

    model_gnn.train()

    epoch_losses = []

    # cada epoch ve distintos snapshots históricos

    random.shuffle(train_snapshots)

    for snapshot in train_snapshots:

        optimizer.zero_grad()

        x = snapshot["x"].to(device)

        edge_index = snapshot["edge_index"].to(device)

        edge_label_index = snapshot["edge_label_index"].to(device)

        y = snapshot["edge_label"].to(device)

        logits = model_gnn(x, edge_index, edge_label_index)

        loss = criterion(logits, y)

        loss.backward()

        optimizer.step()

        epoch_losses.append(loss.item())

    # validación temporal

    val_metrics, _, _ = evaluar_snapshot(model_gnn, val_snapshot)

    mean_loss = np.mean(epoch_losses)

    history.append(
        {
            "Epoch": epoch,
            "Loss": mean_loss,
            "Val_ROC_AUC": val_metrics["ROC_AUC"],
            "Val_PR_AUC": val_metrics["PR_AUC"],
        }
    )

    if val_metrics["ROC_AUC"] > best_val_auc:

        best_val_auc = val_metrics["ROC_AUC"]

        best_state = {k: v.cpu().clone() for k, v in model_gnn.state_dict().items()}

    if epoch == 1 or epoch % 25 == 0:

        print(
            f"Epoch {epoch:3d} | "
            f"Loss {mean_loss:.4f} | "
            f"Val ROC-AUC "
            f"{val_metrics['ROC_AUC']:.3f} | "
            f"Val PR-AUC "
            f"{val_metrics['PR_AUC']:.3f}"
        )


# recuperar mejor modelo

model_gnn.load_state_dict(best_state)

model_gnn = model_gnn.to(device)


# ============================================================
# 9. CURVA DE ENTRENAMIENTO
# ============================================================

history_df = pd.DataFrame(history)


fig, ax1 = plt.subplots(figsize=(11, 6))


ax1.plot(history_df["Epoch"], history_df["Loss"], label="Training loss")

ax1.set_xlabel("Epoch")

ax1.set_ylabel("BCE Loss")


ax2 = ax1.twinx()


ax2.plot(history_df["Epoch"], history_df["Val_ROC_AUC"], label="Validation ROC-AUC")

ax2.plot(
    history_df["Epoch"],
    history_df["Val_PR_AUC"],
    linestyle="--",
    label="Validation PR-AUC",
)

ax2.set_ylabel("Score")


lines1, labels1 = ax1.get_legend_handles_labels()

lines2, labels2 = ax2.get_legend_handles_labels()


ax2.legend(lines1 + lines2, labels1 + labels2, loc="lower right")


plt.title("GraphSAGE temporal — entrenamiento")

plt.tight_layout()
plt.show()


# ============================================================
# 10. EVALUACIÓN FINAL 2024–2026
# ============================================================

val_metrics, _, _ = evaluar_snapshot(model_gnn, val_snapshot)

test_metrics, y_test_gnn, prob_test_gnn = evaluar_snapshot(model_gnn, test_snapshot)


metrics_df = pd.DataFrame(
    [
        {"Dataset": "Validation 2023–2025", **val_metrics},
        {"Dataset": "Test 2024–2026", **test_metrics},
    ]
)


display(metrics_df.round(3))


# ============================================================
# 11. PLOT DE MÉTRICAS
# ============================================================

plot_metrics = metrics_df.melt(id_vars="Dataset", var_name="Metric", value_name="Score")


plt.figure(figsize=(11, 6))


import seaborn as sns

sns.barplot(data=plot_metrics, x="Metric", y="Score", hue="Dataset")


plt.ylim(0, 1)

plt.xlabel("")
plt.ylabel("Score")

plt.title("GraphSAGE — predicción temporal de nuevas colaboraciones")

plt.legend(title="")

plt.tight_layout()
plt.show()


# ============================================================
# 12. MATRIZ DE CONFUSIÓN DEL TEST
# ============================================================

pred_test_gnn = (prob_test_gnn >= 0.5).astype(int)


cm = confusion_matrix(y_test_gnn, pred_test_gnn)


plt.figure(figsize=(6, 5))


sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cbar=False,
    xticklabels=["No colaboración", "Colaboración"],
    yticklabels=["No colaboración", "Colaboración"],
)


plt.xlabel("Predicción")

plt.ylabel("Real")

plt.title("GraphSAGE — Test temporal 2024–2026")

plt.tight_layout()
plt.show()


# ============================================================
# 13. RANKING DE COLABORACIONES PREDICHAS
#     A PARTIR DE LA RED HASTA 2023
# ============================================================

model_gnn.eval()


snapshot = test_snapshot

x = snapshot["x"].to(device)

edge_index = snapshot["edge_index"].to(device)

G_2023 = snapshot["G"]

paper_counts_2023 = snapshot["paper_counts"]


with torch.no_grad():

    z = model_gnn.encode(x, edge_index)


candidate_pairs = []
candidate_names = []


active_authors_2023 = [
    author
    for author in all_authors_gnn
    if paper_counts_2023[author] >= MIN_PAPERS_AUTHOR
]


for i in range(len(active_authors_2023)):

    for j in range(i + 1, len(active_authors_2023)):

        a = active_authors_2023[i]
        b = active_authors_2023[j]

        # solamente pares que nunca colaboraron
        # hasta 2023

        if G_2023.has_edge(a, b):
            continue

        candidate_pairs.append([author_to_idx[a], author_to_idx[b]])

        candidate_names.append((a, b))


candidate_edge_index = (
    torch.tensor(candidate_pairs, dtype=torch.long).T.contiguous().to(device)
)


with torch.no_grad():

    logits = model_gnn.decode(z, candidate_edge_index)

    probabilities = torch.sigmoid(logits).cpu().numpy()


prediction_df = pd.DataFrame(
    {
        "Author1": [x[0] for x in candidate_names],
        "Author2": [x[1] for x in candidate_names],
        "Predicted_Collaboration_Probability": probabilities,
    }
)


prediction_df = prediction_df.sort_values(
    "Predicted_Collaboration_Probability", ascending=False
).reset_index(drop=True)


display(prediction_df.head(30).round(3))


# ============================================================
# 14. TOP PREDICCIONES QUE INVOLUCRAN AUTORES IFIBIO
# ============================================================

ifibio_set = set(ifibio_authors)


prediction_ifibio = prediction_df[
    prediction_df["Author1"].isin(ifibio_set)
    | prediction_df["Author2"].isin(ifibio_set)
].copy()


prediction_ifibio["Pair"] = (
    prediction_ifibio["Author1"] + " — " + prediction_ifibio["Author2"]
)


top_ifibio_gnn = prediction_ifibio.head(30).copy()


top_ifibio_gnn["Probability_Percent"] = (
    top_ifibio_gnn["Predicted_Collaboration_Probability"] * 100
)


display(top_ifibio_gnn[["Author1", "Author2", "Probability_Percent"]].round(1))


plot_data = top_ifibio_gnn.head(20).sort_values("Probability_Percent")


plt.figure(figsize=(11, 9))


sns.barplot(data=plot_data, x="Probability_Percent", y="Pair")


plt.xlabel("Score de probabilidad de nueva colaboración (%)")

plt.ylabel("")

plt.title("GraphSAGE — nuevas colaboraciones potenciales IFIBIO")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# EXPLICAR POR QUÉ DOS AUTORES DEBERÍAN COLABORAR
# usando el GraphSAGE entrenado
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch

# ------------------------------------------------------------
# AUTORES A ANALIZAR
# ------------------------------------------------------------

apellido_1 = "Casal"
apellido_2 = "Di Giusto"


# ------------------------------------------------------------
# BUSCAR AUTORES
# ------------------------------------------------------------

author_1 = [a for a in all_authors_gnn if apellido_1.lower() in a.lower()][0]

author_2 = [a for a in all_authors_gnn if apellido_2.lower() in a.lower()][0]


print("Autor 1:", author_1)
print("Autor 2:", author_2)


# ------------------------------------------------------------
# SNAPSHOT USADO PARA LA PREDICCIÓN
# ------------------------------------------------------------

snapshot = test_snapshot

G_hist = snapshot["G"]

x = snapshot["x"].to(device)

edge_index = snapshot["edge_index"].to(device)


# ------------------------------------------------------------
# EMBEDDINGS GRAPHSAGE
# ------------------------------------------------------------

model_gnn.eval()

with torch.no_grad():

    z = model_gnn.encode(x, edge_index)


i = author_to_idx[author_1]
j = author_to_idx[author_2]


pair_index = torch.tensor([[i], [j]], dtype=torch.long).to(device)


with torch.no_grad():

    logit = model_gnn.decode(z, pair_index)

    probability = torch.sigmoid(logit).cpu().item()


# ------------------------------------------------------------
# HISTORIA DE COLABORACIÓN
# ------------------------------------------------------------

already_collaborated = G_hist.has_edge(author_1, author_2)


# ------------------------------------------------------------
# VECINOS
# ------------------------------------------------------------

neighbors_1 = set(G_hist.neighbors(author_1))

neighbors_2 = set(G_hist.neighbors(author_2))

common_neighbors = neighbors_1 & neighbors_2


# ------------------------------------------------------------
# JACCARD
# ------------------------------------------------------------

union_neighbors = neighbors_1 | neighbors_2

jaccard = (
    len(common_neighbors) / len(union_neighbors) if len(union_neighbors) > 0 else 0
)


# ------------------------------------------------------------
# ADAMIC-ADAR
# ------------------------------------------------------------

aa = list(nx.adamic_adar_index(G_hist, [(author_1, author_2)]))

adamic_adar = aa[0][2] if len(aa) else 0


# ------------------------------------------------------------
# EMBEDDING SIMILARITY
# ------------------------------------------------------------

embedding_1 = z[i].detach().cpu().numpy()

embedding_2 = z[j].detach().cpu().numpy()


embedding_similarity = np.dot(embedding_1, embedding_2) / (
    np.linalg.norm(embedding_1) * np.linalg.norm(embedding_2) + 1e-12
)


# ------------------------------------------------------------
# PRODUCTIVIDAD HISTÓRICA
# ------------------------------------------------------------

papers_1 = snapshot["paper_counts"][author_1]

papers_2 = snapshot["paper_counts"][author_2]


degree_1 = G_hist.degree(author_1)

degree_2 = G_hist.degree(author_2)


weighted_degree_1 = G_hist.degree(author_1, weight="weight")

weighted_degree_2 = G_hist.degree(author_2, weight="weight")


# ------------------------------------------------------------
# TABLA RESUMEN
# ------------------------------------------------------------

explanation = pd.DataFrame(
    {
        "Métrica": [
            "Score GraphSAGE",
            "Similitud embedding",
            "Colaboradores comunes",
            "Jaccard",
            "Adamic-Adar",
            f"Publicaciones {author_1}",
            f"Publicaciones {author_2}",
            f"Colaboradores {author_1}",
            f"Colaboradores {author_2}",
        ],
        "Valor": [
            probability,
            embedding_similarity,
            len(common_neighbors),
            jaccard,
            adamic_adar,
            papers_1,
            papers_2,
            degree_1,
            degree_2,
        ],
    }
)


display(explanation.round(3))


print("\n¿Ya colaboraron?:", already_collaborated)


print("\nColaboradores comunes:")

for author in sorted(common_neighbors):

    print(" •", author)


# ------------------------------------------------------------
# PERFIL COMPARATIVO
# mismas variables utilizadas por GraphSAGE
# ------------------------------------------------------------

feature_names_gnn = [
    "Publicaciones",
    "Papers último año",
    "Papers últimos 3 años",
    "Longitud carrera",
    "Degree",
    "Weighted degree",
    "PageRank",
    "Clustering",
]


profile = pd.DataFrame(
    {
        "Feature": feature_names_gnn,
        author_1: snapshot["x"][i].numpy(),
        author_2: snapshot["x"][j].numpy(),
    }
)


profile_long = profile.melt(
    id_vars="Feature", var_name="Author", value_name="Valor estandarizado"
)


plt.figure(figsize=(11, 7))


sns.barplot(data=profile_long, y="Feature", x="Valor estandarizado", hue="Author")


plt.axvline(0, linewidth=1, linestyle="--")


plt.xlabel("Valor estandarizado")

plt.ylabel("")

plt.title(f"¿Por qué {author_1} y {author_2} " "son compatibles según GraphSAGE?")

plt.legend(title="")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# RESUMEN AUTOMÁTICO
# ------------------------------------------------------------

print("\n==============================")
print("INTERPRETACIÓN")
print("==============================")

print(
    f"\nGraphSAGE asigna al par "
    f"{author_1} — {author_2} "
    f"un score de {probability*100:.1f}%."
)

print(f"\nSimilitud entre embeddings: " f"{embedding_similarity:.3f}")

print(f"Colaboradores comunes: " f"{len(common_neighbors)}")

print(f"Jaccard: " f"{jaccard:.3f}")

print(f"Adamic-Adar: " f"{adamic_adar:.3f}")

if len(common_neighbors) > 0:

    print(
        "\nLa pareja ocupa regiones conectadas "
        "de la red científica y comparte "
        "intermediarios potenciales."
    )

if embedding_similarity > 0.5:

    print(
        "\nGraphSAGE aprendió representaciones "
        "relativamente similares para ambos autores."
    )

if not already_collaborated:

    print(
        "\nNo existe colaboración previa en el "
        "snapshot histórico: constituye un enlace "
        "potencialmente nuevo."
    )
```
