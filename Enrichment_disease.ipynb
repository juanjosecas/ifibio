```python
import pandas as pd

pubtator_entities = pd.read_csv("pubtator3_entities_all.csv")

gprofiler_enrichment = pd.read_csv("ifibio_gprofiler_enrichment.csv")

display(pubtator_entities.head())
display(gprofiler_enrichment.head())

print(pubtator_entities.columns)
print(gprofiler_enrichment.columns)

print("Examples of the data:")

max_rows = 5
print("PubTator Entities:")
for index, row in pubtator_entities.iterrows():
    if index >= max_rows:
        break
    print(f"Row {index}: {row.to_dict()}")

print("\nGProfiler Enrichment:")
for index, row in gprofiler_enrichment.iterrows():
    if index >= max_rows:
        break
    print(f"Row {index}: {row.to_dict()}")
```

```python
# ============================================================
# ANÁLISIS FUNCIONAL A PARTIR DE LOS CSV GUARDADOS
# Corrección de Fold Enrichment + métricas informativas
# ============================================================

import ast
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

enr = gprofiler_enrichment.copy()


# ------------------------------------------------------------
# 1. Recalcular métricas correctamente
# ------------------------------------------------------------

# Gene ratio / precision
enr["GeneRatio"] = enr["intersection_size"] / enr["query_size"]

# Background frequency del término
enr["BackgroundRatio"] = enr["term_size"] / enr["effective_domain_size"]

# Fold enrichment REAL
enr["FoldEnrichment"] = enr["GeneRatio"] / enr["BackgroundRatio"]

# Significancia
enr["minus_log10_p"] = -np.log10(enr["p_value"])

# Especificidad aproximada:
# términos pequeños = más específicos
enr["Specificity"] = 1 / enr["term_size"]


# ------------------------------------------------------------
# 2. Parsear padres GO/pathway
# ------------------------------------------------------------


def parse_list(x):
    if isinstance(x, list):
        return x

    if pd.isna(x):
        return []

    try:
        value = ast.literal_eval(str(x))

        if isinstance(value, list):
            return value

    except:
        pass

    return []


enr["ParentList"] = enr["parents"].apply(parse_list)


# ------------------------------------------------------------
# 3. Resumen global
# ------------------------------------------------------------

print("Términos totales:", len(enr))

print("Términos significativos:", enr["significant"].sum())

print("Fuentes:", sorted(enr["source"].unique()))


source_summary = (
    enr.groupby("source")
    .agg(
        Terms=("native", "nunique"),
        Significant=("significant", "sum"),
        Median_FDR=("p_value", "median"),
        Median_FE=("FoldEnrichment", "median"),
        Max_FE=("FoldEnrichment", "max"),
        Median_term_size=("term_size", "median"),
    )
    .sort_values("Significant", ascending=False)
)

display(source_summary)


# ------------------------------------------------------------
# 4. Términos más enriquecidos
# ------------------------------------------------------------

top_fe = enr[enr["significant"] == True].sort_values(
    ["FoldEnrichment", "p_value"], ascending=[False, True]
)[
    [
        "source",
        "native",
        "name",
        "p_value",
        "intersection_size",
        "term_size",
        "GeneRatio",
        "BackgroundRatio",
        "FoldEnrichment",
    ]
]

print("\nTÉRMINOS CON MAYOR FOLD ENRICHMENT")

display(top_fe.head(30))


# ------------------------------------------------------------
# 5. Términos más significativos pero suficientemente
#    específicos
#
# Evitamos términos GO enormes y genéricos
# ------------------------------------------------------------

specific_terms = enr[
    (enr["significant"] == True) & (enr["term_size"] >= 10) & (enr["term_size"] <= 1000)
].sort_values("p_value")

print("\nTÉRMINOS SIGNIFICATIVOS Y MÁS ESPECÍFICOS")

display(
    specific_terms[
        [
            "source",
            "native",
            "name",
            "p_value",
            "intersection_size",
            "term_size",
            "FoldEnrichment",
        ]
    ].head(40)
)


# ============================================================
# PLOT 1
# DOTPLOT BIOLÓGICO
#
# x      = Fold enrichment
# size   = genes observados
# color  = significancia
# ============================================================

plot = specific_terms.head(30).copy()

plot["Label"] = plot["name"].astype(str).str.slice(0, 65)

plot = plot.sort_values("FoldEnrichment")

fig, ax = plt.subplots(figsize=(10, 8))

sc = ax.scatter(
    plot["FoldEnrichment"],
    plot["Label"],
    s=plot["intersection_size"] * 8,
    c=plot["minus_log10_p"],
    cmap="viridis",
    alpha=0.85,
    edgecolors="black",
    linewidths=0.4,
)

ax.set_xlabel("Fold enrichment")
ax.set_ylabel("")
ax.set_title("Functional enrichment of IFIBIO research genes")

cbar = plt.colorbar(sc, ax=ax)

cbar.set_label("-log10(FDR)")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 2
# SIGNIFICANCIA vs ESPECIFICIDAD
#
# Permite separar términos enormes/generalistas
# de términos biológicamente más concretos
# ============================================================

plot2 = enr[(enr["significant"] == True) & (enr["term_size"] > 0)].copy()

fig, ax = plt.subplots(figsize=(8, 6))

for source, sub in plot2.groupby("source"):

    ax.scatter(sub["term_size"], sub["minus_log10_p"], alpha=0.5, s=25, label=source)

ax.set_xscale("log")

ax.set_xlabel("Term size (log scale)")
ax.set_ylabel("-log10(FDR)")

ax.set_title("Significance versus functional term size")

ax.legend(frameon=False, fontsize=8)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 3
# FOLD ENRICHMENT vs SIGNIFICANCIA
#
# Muy útil para detectar señales fuertes,
# no sólo términos con p pequeño
# ============================================================

fig, ax = plt.subplots(figsize=(8, 6))

for source, sub in plot2.groupby("source"):

    ax.scatter(
        sub["FoldEnrichment"], sub["minus_log10_p"], alpha=0.55, s=25, label=source
    )

ax.set_xlabel("Fold enrichment")
ax.set_ylabel("-log10(FDR)")

ax.set_title("Strength and significance of functional enrichment")

ax.legend(frameon=False, fontsize=8)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. Guardar tabla corregida
# ------------------------------------------------------------

enr.to_csv("ifibio_gprofiler_enrichment_corrected.csv", index=False)
```

```python
# ============================================================
# DISEASE LANDSCAPE — PubTator3 / IFIBIO
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Filtrar enfermedades válidas
# ------------------------------------------------------------

diseases = pubtator_entities[
    (pubtator_entities["EntityType"] == "Disease")
    & (pubtator_entities["Valid"] == True)
    & (pubtator_entities["NormalizedID"].notna())
].copy()

diseases["PMID"] = diseases["PMID"].astype(str)

print("Anotaciones válidas de enfermedad:", len(diseases))
print("Papers con enfermedades:", diseases["PMID"].nunique())
print("Enfermedades normalizadas únicas:", diseases["NormalizedID"].nunique())


# ------------------------------------------------------------
# 2. Catálogo de enfermedades
# ------------------------------------------------------------

disease_catalog = (
    diseases.groupby(["NormalizedID", "NormalizedName"], dropna=False)
    .agg(
        Papers=("PMID", "nunique"),
        Mentions=("Mention", "size"),
        Text_mentions=("Mention", lambda x: sorted(set(x.dropna().astype(str)))),
        Sections=("Section", lambda x: sorted(set(x.dropna().astype(str)))),
    )
    .reset_index()
    .sort_values(["Papers", "Mentions"], ascending=False)
)

print("\nENFERMEDADES MÁS FRECUENTES")
display(disease_catalog.head(40))


# ------------------------------------------------------------
# 3. Enfermedades por paper
# ------------------------------------------------------------

diseases_per_paper = diseases.groupby("PMID")["NormalizedID"].nunique()

print("\nENFERMEDADES ÚNICAS POR PAPER")

display(diseases_per_paper.describe().to_frame("N_diseases"))


# ------------------------------------------------------------
# 4. Título vs abstract
# ------------------------------------------------------------

section_summary = (
    diseases.groupby("Section")
    .agg(
        Mentions=("Mention", "size"),
        Papers=("PMID", "nunique"),
        Diseases=("NormalizedID", "nunique"),
    )
    .sort_values("Mentions", ascending=False)
)

print("\nDISTRIBUCIÓN POR SECCIÓN")
display(section_summary)


# ------------------------------------------------------------
# 5. Variantes textuales / aliases
# ------------------------------------------------------------

disease_aliases = (
    diseases.groupby(["NormalizedID", "NormalizedName"])
    .agg(
        N_aliases=("Mention", "nunique"),
        Aliases=("Mention", lambda x: sorted(set(x.dropna().astype(str)))),
        Papers=("PMID", "nunique"),
    )
    .reset_index()
    .sort_values(["Papers", "N_aliases"], ascending=False)
)

print("\nVARIANTES TEXTUALES")

display(disease_aliases.head(30))


# ------------------------------------------------------------
# 6. Ambigüedad:
#    misma mención textual → varios IDs MeSH
# ------------------------------------------------------------

mention_map = (
    diseases.assign(
        Mention_clean=lambda x: x["Mention"].astype(str).str.lower().str.strip()
    )
    .groupby("Mention_clean")
    .agg(
        N_IDs=("NormalizedID", "nunique"),
        Disease_IDs=("NormalizedID", lambda x: sorted(set(map(str, x)))),
        Disease_names=("NormalizedName", lambda x: sorted(set(x.dropna().astype(str)))),
        Papers=("PMID", "nunique"),
    )
    .reset_index()
)

disease_ambiguity = mention_map[mention_map["N_IDs"] > 1].sort_values(
    ["Papers", "N_IDs"], ascending=False
)

print("\nMENCIONES POTENCIALMENTE AMBIGUAS")
display(disease_ambiguity.head(30))


# ============================================================
# PLOT 1
# Enfermedades más presentes en publicaciones
# ============================================================

plot = disease_catalog.head(25).sort_values("Papers")

fig, ax = plt.subplots(figsize=(9, 7))

ax.barh(plot["NormalizedName"], plot["Papers"])

ax.set_xlabel("Number of papers")
ax.set_ylabel("")
ax.set_title("Diseases represented in IFIBIO publications")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 2
# Papers vs número total de menciones
# ============================================================

plot = disease_catalog.copy()

fig, ax = plt.subplots(figsize=(7.5, 5.5))

ax.scatter(plot["Papers"], plot["Mentions"], alpha=0.6, s=35)

top_labels = plot.sort_values(["Papers", "Mentions"], ascending=False).head(15)

for _, row in top_labels.iterrows():

    ax.annotate(
        str(row["NormalizedName"]),
        (row["Papers"], row["Mentions"]),
        xytext=(4, 4),
        textcoords="offset points",
        fontsize=8,
    )

ax.set_xlabel("Number of papers")
ax.set_ylabel("Number of mentions")

ax.set_title("Disease recurrence across IFIBIO publications")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 3
# Distribución de enfermedades por paper
# ============================================================

fig, ax = plt.subplots(figsize=(7, 4.5))

bins = np.arange(diseases_per_paper.min(), diseases_per_paper.max() + 2) - 0.5

ax.hist(diseases_per_paper, bins=bins)

ax.set_xlabel("Unique diseases per paper")
ax.set_ylabel("Number of papers")

ax.set_title("Disease annotation density")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. Guardar
# ------------------------------------------------------------

disease_catalog.to_csv("ifibio_disease_catalog.csv", index=False)

disease_aliases.to_csv("ifibio_disease_aliases.csv", index=False)

disease_ambiguity.to_csv("ifibio_disease_ambiguity.csv", index=False)
```

```python
# ============================================================
# GENE ↔ DISEASE CO-MENTION NETWORK
# PubTator3 / IFIBIO
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

# ------------------------------------------------------------
# 1. Genes válidos
# ------------------------------------------------------------

genes = pubtator_entities[
    (pubtator_entities["EntityType"] == "Gene")
    & (pubtator_entities["Valid"] == True)
    & (pubtator_entities["NormalizedID"].notna())
].copy()

genes["PMID"] = genes["PMID"].astype(str)
genes["NormalizedID"] = genes["NormalizedID"].astype(str)


# ------------------------------------------------------------
# 2. Enfermedades válidas
# ------------------------------------------------------------

diseases = pubtator_entities[
    (pubtator_entities["EntityType"] == "Disease")
    & (pubtator_entities["Valid"] == True)
    & (pubtator_entities["NormalizedID"].notna())
    & (pubtator_entities["NormalizedName"].notna())
].copy()

diseases["PMID"] = diseases["PMID"].astype(str)


# ------------------------------------------------------------
# 3. Una entidad por paper
# ------------------------------------------------------------

paper_genes = (
    genes[["PMID", "NormalizedID", "NormalizedName"]]
    .drop_duplicates()
    .rename(columns={"NormalizedID": "GeneID", "NormalizedName": "Gene"})
)

paper_diseases = (
    diseases[["PMID", "NormalizedID", "NormalizedName"]]
    .drop_duplicates()
    .rename(columns={"NormalizedID": "DiseaseID", "NormalizedName": "Disease"})
)


# ------------------------------------------------------------
# 4. Combinar dentro de cada paper
# ------------------------------------------------------------

gene_disease_raw = paper_genes.merge(paper_diseases, on="PMID", how="inner")

print("Combinaciones Gene-Disease:", len(gene_disease_raw))

print("Papers con genes + enfermedades:", gene_disease_raw["PMID"].nunique())


# ------------------------------------------------------------
# 5. Número de papers que soportan cada asociación
# ------------------------------------------------------------

gene_disease = (
    gene_disease_raw.groupby(["GeneID", "Gene", "DiseaseID", "Disease"])
    .agg(Papers=("PMID", "nunique"), PMIDs=("PMID", lambda x: sorted(set(x))))
    .reset_index()
    .sort_values("Papers", ascending=False)
)


print("\nASOCIACIONES MÁS RECURRENTES")

display(gene_disease.head(40))


# ============================================================
# 6. Genes con mayor diversidad de enfermedades
# ============================================================

gene_disease_profile = (
    gene_disease.groupby(["GeneID", "Gene"])
    .agg(
        Diseases=("DiseaseID", "nunique"),
        Associations=("DiseaseID", "size"),
        Supporting_papers=("Papers", "sum"),
    )
    .reset_index()
    .sort_values(["Diseases", "Supporting_papers"], ascending=False)
)

print("\nGENES ASOCIADOS A MAYOR DIVERSIDAD DE ENFERMEDADES")

display(gene_disease_profile.head(30))


# ============================================================
# 7. Enfermedades con mayor diversidad de genes
# ============================================================

disease_gene_profile = (
    gene_disease.groupby(["DiseaseID", "Disease"])
    .agg(
        Genes=("GeneID", "nunique"),
        Associations=("GeneID", "size"),
        Supporting_papers=("Papers", "sum"),
    )
    .reset_index()
    .sort_values(["Genes", "Supporting_papers"], ascending=False)
)

print("\nENFERMEDADES CON MAYOR DIVERSIDAD DE GENES")

display(disease_gene_profile.head(30))


# ============================================================
# 8. PLOT — enfermedades con más genes
# ============================================================

plot = disease_gene_profile.head(25).sort_values("Genes")

fig, ax = plt.subplots(figsize=(9, 7))

ax.barh(plot["Disease"], plot["Genes"])

ax.set_xlabel("Number of genes")
ax.set_ylabel("")

ax.set_title("Diseases associated with the largest gene diversity")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()


# ============================================================
# 9. PLOT — genes con más enfermedades
# ============================================================

plot = gene_disease_profile.head(25).sort_values("Diseases")

fig, ax = plt.subplots(figsize=(8, 6))

ax.barh(plot["Gene"], plot["Diseases"])

ax.set_xlabel("Number of diseases")
ax.set_ylabel("")

ax.set_title("Genes associated with the largest disease diversity")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()


# ============================================================
# 10. RED BIPARTITA
#
# Nos quedamos inicialmente con asociaciones respaldadas
# por >= 2 papers para reducir ruido.
# ============================================================

network_edges = gene_disease[gene_disease["Papers"] >= 2].copy()

print("\nEdges con >=2 papers:", len(network_edges))


G = nx.Graph()


for _, row in network_edges.iterrows():

    gene_node = f"G::{row['Gene']}"
    disease_node = f"D::{row['Disease']}"

    G.add_node(gene_node, label=row["Gene"], node_type="gene")

    G.add_node(disease_node, label=row["Disease"], node_type="disease")

    G.add_edge(gene_node, disease_node, weight=row["Papers"])


print("Nodos:", G.number_of_nodes())

print("Edges:", G.number_of_edges())


# ------------------------------------------------------------
# 11. Componentes
# ------------------------------------------------------------

components = sorted(nx.connected_components(G), key=len, reverse=True)

print("Componentes conectados:", len(components))

if components:
    print("Mayor componente:", len(components[0]), "nodos")


# ------------------------------------------------------------
# 12. Plot
# ------------------------------------------------------------

if G.number_of_nodes() > 0:

    pos = nx.spring_layout(G, seed=42, k=1.0, weight="weight")

    gene_nodes = [n for n, d in G.nodes(data=True) if d["node_type"] == "gene"]

    disease_nodes = [n for n, d in G.nodes(data=True) if d["node_type"] == "disease"]

    fig, ax = plt.subplots(figsize=(13, 10))

    nx.draw_networkx_edges(
        G,
        pos,
        width=[0.5 + G[u][v]["weight"] * 0.7 for u, v in G.edges()],
        alpha=0.25,
        ax=ax,
    )

    nx.draw_networkx_nodes(
        G, pos, nodelist=gene_nodes, node_size=180, node_shape="o", alpha=0.85, ax=ax
    )

    nx.draw_networkx_nodes(
        G, pos, nodelist=disease_nodes, node_size=350, node_shape="s", alpha=0.75, ax=ax
    )

    labels = {n: G.nodes[n]["label"] for n in G.nodes()}

    nx.draw_networkx_labels(G, pos, labels=labels, font_size=7, ax=ax)

    ax.set_title("Gene–disease landscape in IFIBIO publications")

    ax.axis("off")

    plt.tight_layout()
    plt.show()


# ============================================================
# 13. Guardar
# ============================================================

gene_disease.to_csv("ifibio_gene_disease_comention.csv", index=False)

gene_disease_profile.to_csv("ifibio_gene_disease_profile.csv", index=False)

disease_gene_profile.to_csv("ifibio_disease_gene_profile.csv", index=False)
```
